## Нативные инструменты Python

### venv (Встроенный модуль)

1. **Шаг 1.** Создание виртуального окружения

```bash
# Создать venv в папке .venv (стандартное имя)
python -m venv .venv

# Или с конкретной версией Python
python3.11 -m venv .venv
```

2. **Шаг 2.** Активация venv

Windows:
```bash
.venv\Scripts\activate
```

macOS/Linux:
```
source .venv/bin/activate
```

3. **Шаг 3.** Проверка активации

```bash
# Появится (.venv) перед строкой приглашения
which python    # Linux/macOS — покажет путь внутри venv
where python    # Windows
```

4. **Шаг 4.** Управление пакетами

Установка пакетов:

```bash
pip install requests
pip install django flask
```

Создание requirements.txt:

```bash
# Экспорт всех установленных пакетов
pip freeze > requirements.txt
```

Установка из requirements.txt:
```bash
pip install -r requirements.txt
```

5. Деактивация

```bash
deactivate
```

6. Удаление venv

```bash
# Просто удалить папку
rm -rf .venv   # Linux/macOS
rmdir /s .venv  # Windows
```

Файл dependencies.txt (пример)

```txt
# Обычно называют requirements.txt
django==4.2.7
requests==2.31.0
python-dotenv>=1.0.0
pytest~=7.4.0
```



Спецификаторы версий:

- `==` — точная версия

- `>=` — не ниже

- `~=` — совместимая версия (`7.4.0` ≤ версия < `8.0.0`)

Полезные команды

```bash
# Просмотр установленных пакетов
pip list

# Проверка устаревших пакетов
pip list --outdated

# Обновление pip
python -m pip install --upgrade pip
```

## Conda: Экосистема, а не только окружение

### 0. Установка `Conda`
Часто Conda не представлена в официальных репозиториях. Например в Arch Conda есть только в репозитории AUR.
Поэтому Conda лучше загрузить вручную по следующей ссылке: https://repo.anaconda.com/archive/. Например:

```bash
curl -O https://repo.anaconda.com/archive/Anaconda3-2024.06-1-Linux-x86_64.sh
bash Anaconda3-2024.06-1-Linux-x86_64.sh
```

Далее запускаем скрипт `Anaconda3-2024.06-1-Linux-x86_64.sh`, он оуществит процесс установки.

### 1. Создание окружения

```bash
# Создать окружение с Python 3.11
conda create -n llmsim-env python=3.11

# С указанием конкретных пакетов
conda create -n llmsim-env python=3.11 numpy pandas

# С указанием канала (например, conda-forge)
conda create -n llmsim-env -c conda-forge python=3.11
```

### 2. Активация окружения

```bash
conda activate llmsim-env
```

### 3. Использование окружения как ядро для JupyterLab:

Создайте и активируйте ваше Conda-окружение (если вы этого еще не сделали):
```bash
conda create -n имя_окружения python=3.11
conda activate имя_окружения
```

Установите ipykernel в это окружение:

```bash
conda install ipykernel
```

Или через pip, если предпочитаете: `pip install ipykernel`.

Зарегистрируйте окружение как ядро для Jupyter:

```bash
python -m ipykernel install --user --name=имя_окружения --display-name="Имя, которое вы увидите в Jupyter"
```

- `--name`: это внутреннее техническое имя ядра.
- `--display-name`: это имя, которое отобразится в лаунчере JupyterLab.

После этого обновите страницу JupyterLab в браузере — ваше новое ядро должно появиться в списке доступных.

### 4. Файл environment.yml:

```yml
name: my_project
channels:
  - conda-forge   # Самый популярный канал с новыми версиями
  - defaults
dependencies:
  - python=3.11
  - numpy=1.24
  - pandas
  - pip           # Обязательно добавляем pip, если будем его использовать
  - pip:
    - requests    # Это установится через pip, т.к. в conda-forge его может не быть последней версии
```

Использование:

```bash
conda env create -f environment.yml
```
