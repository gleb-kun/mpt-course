It is necessary to check whether the inequalities $y(-x)=-y(x)$ and $y(-x)=y(x)$ are satisfied.

- Find the intersection points with the coordinate axes and the intervals of constant sign;
Solve the corresponding equations.

- Find the asymptotes of the function's graph;
The equation of the oblique asymptote is $y=kx+b,$ where the coefficients are calculated using the formulas:
$$
k=\lim_{x\rightarrow +\infty}\frac{f(x)}{x}\quad\text{and}\quad b=\lim_{x\rightarrow\infty}(f(x)-kx).\tag{1}
$$
Similar formulas for $x\rightarrow-\infty$ (usually the coefficients are the same).

**Description of the problem:**

If a surface is defined by the equation
$$F(x, y, z) = 0\tag{2},$$
that is, implicitly, then the equation of the tangent plane to this surface at the point
$M_0(x_0, y_0, z_0)$ can be found by the following formula:
$$F'_x(M_0)(x - x_0) + F'_y(M_0)(y - y_0) + F'_z(M_0)(z - z_0) = 0.\tag{3}$$

The equation of the normal at the point $M_0(x_0,y_0,z_0)$:
$$\frac{(x - x_0)}{F'_x(M_0)} = \frac{(y - y_0)}{F'_y(M_0)} = \frac{(z - z_0)}{F'_z(M_0)}.\tag{4}$$
