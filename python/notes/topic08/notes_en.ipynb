## Planning a Numerical Experiment

Planning a numerical experiment can be represented as the following sequential steps:

- selecting test functions,
- choosing a method,
- choosing a step,
- setting criteria for comparing the result and the exact solution, and verifying the result.

### Selecting Test Functions

Test functions must have a known exact value to be able to calculate the error.

| Function | Derivative | Integral on [0,1] | Singularity |
|---|---|---|---|
| $f(x) = x^2$ | $ 2x $ | $ 1/3 $ | Smooth, polynomial |
| $f(x) = x^3$ | $ 3x^2 $ | $ 1/4 $ | Smooth |
| $f(x) = e^x $ | $ e^x $ | $ e - 1 $ | Exponential |
| $ f(x) = \sin(x) $ | $ \cos(x) $ | $ 1 - \cos(1) $ | Trigonometric |
| $ f(x) = \sqrt{x} $ | $ 1/(2\sqrt{x}) $ | $ 2/3 $ | Nonsmooth at zero |
| $ f(x) = x^4 \sin(1/x) $ | — | — | Strongly oscillatory |

### Selecting Methods

The following well-known numerical methods can be considered for comparison.

| Problem | Methods to Compare |
|---|---|
| **Differentiation** | Right Difference, Left Difference, Central Difference |
| **Integration** | Rectangle Formula (Left/Right/Middle), Trapezoidal Formula, Simpson's Formula |
| **Solving ODEs** | Euler's Method (Explicit/Implicit), Runge-Kutta Method of 2nd and 4th Order |

#### Brief Description of Numerical Methods

##### Numerical Differentiation

1. Left Difference
<br>
$$
f'(x_0) \approx \frac{f(x_0) - f(x_0 - h)}{h}
$$
<br>
Order of accuracy: $ O(h)$.

2. Right Difference
<br>
$$
f'(x_0) \approx \frac{f(x_0 + h) - f(x_0)}{h}
$$
<br>
Order of accuracy: $ O(h)$.

3. **Central Difference**
<br>
$$
f'(x_0) \approx \frac{f(x_0 + h) - f(x_0 - h)}{2h}
$$
<br>
Order of accuracy: $ O(h^2) $.

##### Numerical Integration (Composite Quadrature Formulas)

1. **Composite Left Rectangle Formula**
<br>
$$
\int_a^b f(x) \, dx \approx h \sum_{i=0}^{n-1} f(x_i), \quad h = \frac{b-a}{n}, \quad x_i = a + i \cdot h
$$
<br>
Order of accuracy: $ O(h) $.

2. **Right Rectangle Composite Formula**
<br>
$$
\int_a^b f(x) \, dx \approx h \sum_{i=1}^{n} f(x_i), \quad h = \frac{b-a}{n}, \quad x_i = a + i \cdot h
$$
<br>
Order of accuracy: $ O(h) $.

3. **Midpoint Rectangle Formula**
<br>
$$
\int_a^b f(x) \, dx \approx h \sum_{i=0}^{n-1} f\left( \frac{x_i + x_{i+1}}{2} \right), \quad h = \frac{b-a}{n}
$$
<br>
Order of accuracy: $ O(h^2) $.

4. **Trapezoidal Rule**
<br>
$$
\int_a^b f(x) \, dx \approx h \left( \frac{f(x_0)}{2} + \sum_{i=1}^{n-1} f(x_i) + \frac{f(x_n)}{2} \right), \quad h = \frac{b-a}{n}
$$
<br>
Order of accuracy: $ O(h^2) $.

5. **Simpson's Rule**
<br>
$$
\int_a^b f(x) \, dx \approx \frac{h}{3} \left( f(x_0) + 4\sum_{i \text{ odd}} f(x_i) + 2\sum_{i \text{ even}} f(x_i) + f(x_n) \right), \quad h = \frac{b-a}{n}, \quad n \text{ is even}
$$
<br>
Order of accuracy: $ O(h^4) $

##### Numerical solution of ODE

1. **Explicit Euler method**
<br>
Problem: $ y' = f(t, y), \quad y(t_0) = y_0 $
<br>
$$
y_{n+1} = y_n + h \cdot f(t_n, y_n)
$$
<br>
Order of accuracy: $ O(h) $

2. **Implicit Euler Method**
<br>
$$
y_{n+1} = y_n + h \cdot f(t_{n+1}, y_{n+1})
$$
<br>
Order of accuracy: $ O(h) $.<br>
**Note:** A nonlinear equation must be solved at each step.

4. **Runge-Kutta Method of the Second Order (RK2 / Heun's Method)**
<br>
$$
\begin{aligned}
k_1 &= f(t_n, y_n) \\
k_2 &= f(t_n + h, y_n + h \cdot k_1) \\
y_{n+1} &= y_n + \frac{h}{2} (k_1 + k_2)
\end{aligned}
$$<br>
Order of accuracy: $ O(h^2) $.

5. **Runge-Kutta Method of the 4th Order (RK4)**
<br>
$$
\begin{aligned}
k_1 &= f(t_n, y_n) \\
k_2 &= f(t_n + \frac{h}{2}, y_n + \frac{h}{2} k_1) \\
k_3 &= f(t_n + \frac{h}{2}, y_n + \frac{h}{2} k_2) \\
k_4 &= f(t_n + h, y_n + h k_3) \\
y_{n+1} &= y_n + \frac{h}{6} (k_1 + 2k_2 + 2k_3 + k_4)
\end{aligned}
$$<br>
Order of accuracy: $ O(h^4)$.

### Step Selection

This example considers grid methods.

Based on this, the step $ h $ should decrease sequentially:

$$
h = 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001, \ldots
$$

Or in powers of 10:

$$
h = 10^{-1}, 10^{-2}, 10^{-3}, \ldots, 10^{-15}.
$$

### Comparison Criteria

| Criterion | Description |
|---|---|
| **Actual Error** | $ \varphi(h) = \| \text{exact solution} - \text{approximate solution} \| $ |
| **Order of Accuracy** | $ p = \lim_{h \to 0} \frac{\log \varphi(h)}{\log h} $ |
| **Constant** | $ C = \lim_{h \to 0} \frac{\varphi(h)}{h^p} $ |
| **Stability** | Error Behavior at Very Small $ h $ |

## Experimental Data Collection and Processing

### Results Form

| $ h $ | $ \varphi(h) $ | $ \lg h $ | $ \lg \varphi(h) $ | $ p_{\text{exp}} $ |
|---|---|---|---|---|
| $ 10^{-1} $ | $ 0.1234 $ | -1.000 | -0.909 | — |
| $ 10^{-2} $ | $ 0.0123 $ | -2.000 | -1.910 | 1.001 |
| $ 10^{-3} $ | $ 0.0012 $ | -3.000 | -2.911 | 1.001 |
| ... | ... | ... | ... | ... |

### Calculating the experimental order of accuracy

For two consecutive steps $ h_1 $ and $ h_2 $:

$$
p_{\text{exp}} = \frac{\log(\varphi(h_1) / \varphi(h_2))}{\log(h_1 / h_2)}.
$$

If $ h_1 = 2h_2 $, then the formula is simplified:

$$
p_{\text{exp}} = \log_2 \left( \frac{\varphi(2h)}{\varphi(h)} \right).
$$

### Estimating the constant $ C $

$$
C \approx \frac{\varphi(h)}{h^p}.
$$

The average is taken over several values ​​of $ h $.

## Results Visualization

### lg-lg graphs

Plotted using pairs of $ (\lg h, \lg \varphi(h)) $.

**Properties:**
- If $ \varphi(h) \approx C \cdot h^p $, then the graph is a straight line with slope $ p $.
- The absolute term $ a = \lg C $.

### Zones on an lg-lg graph

| Zone | Description |
|---|---|
| **Zone 1** | $ h $ is large, the method does not work, the error does not follow the law $ C \cdot h^p $ |
| **Zone 2** | Asymptotic zone, $ \varphi(h) \approx C \cdot h^p $, the graph is a straight line |
| **Zone 3** | Rounding error dominates, the graph "bends" upward |

### Optimal step

Point of minimum total error:

$$
\varphi_{\text{total}}(h) = \varphi_{\text{approx}}(h) + \varphi_{\text{round}}(h).
$$

$ h_{\text{opt}} $ is the step at which $ \varphi_{\text{total}}(h) $ is minimal.

## Stability Analysis

### Impact of Rounding Errors

For very small $ h $, catastrophic cancellation occurs—subtraction of close numbers.

**Example for the right-hand difference:**

$$
\varphi_{\text{rounded}} \approx \frac{2\varepsilon}{h} \cdot |f(x_0)|,
$$

where $ \varepsilon \approx 10^{-16} $ is the machine epsilon.

```python
eps = 1.0
while 1.0 + eps != 1.0:
    eps /= 2
print(eps * 2)
```

### Total Error

$$
\varphi_{\text{total}}(h) \approx C \cdot h^p + \frac{D}{h}
$$

where $ D $ is a constant depending on $ f(x_0) $ and $ \varepsilon $.

### Optimal Step (Analytical Estimate)

From the condition $ \frac{d}{dh} \varphi_{\text{total}}(h) = 0 $:

$$
h_{\text{opt}} = \left( \frac{D}{p \cdot C} \right)^{1/(p+1)}
$$

## Comparison of Methods

### Comparison by Order of Accuracy

| Method | Order $ p $ | Comment |
|---|---|---|
| Left/Right Difference | 1 | Slow Convergence |
| Central Difference | 2 | Good Choice for Smooth Functions |
| Trapezoid Method | 2 | For Integration |
| Simpson's Method | 4 | Very Accurate for Smooth Functions |

### Comparison by Constant $ C $

For the same $ p $, the method with the smaller $ C $ is more accurate.

**Example** (integration by the rectangle method):

| Method | $ C $ | Comment |
|---|---|---|
| Left Rectangle Formula | $ 1/2 $ | Worse |
| Right Rectangle Formula | $ 1/2 $ | Worse |
| Middle Rectangle Formula | $ 1/24 $ | Better |

### Stability Comparison

| Method | Stability at Small $ h $ |
|---|---|
| Left/Right Difference | Moderate |
| Central Difference | Better (roundoff error $ O(1/h^2) $) |
| 4th-Order Runge-Kutta Method | Good for **soft** problems |
