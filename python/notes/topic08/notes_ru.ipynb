# Тема #8. Численные эксперименты и анализ вычислительных ошибок

## Планирование численного эксперимента

Планирование численного эксперимента можно представить в виде следующих последовательных пунктов:

- выбор тестовых функций,
- выбор метода,
- выбор шага,
- постановка критериев сравнения результата и точного решения, верификация результата.

### Выбор тестовых функций

Тестовые функции должны иметь известное точное значение, чтобы можно было вычислить погрешность.

| Функция | Производная | Интеграл на [0,1] | Особенность |
|---|---|---|---|
| $f(x) = x^2$ | $ 2x $ | $ 1/3 $ | Гладкая, полиномиальная |
| $f(x) = x^3$ | $ 3x^2 $ | $ 1/4 $ | Гладкая |
| $ f(x) = e^x $ | $ e^x $ | $ e - 1 $ | Экспоненциальная |
| $ f(x) = \sin(x) $ | $ \cos(x) $ | $ 1 - \cos(1) $ | Тригонометрическая |
| $ f(x) = \sqrt{x} $ | $ 1/(2\sqrt{x}) $ | $ 2/3 $ | Негладкая в нуле |
| $ f(x) = x^4 \sin(1/x) $ | — | — | Сильно осциллирующая |

### Выбор методов

Для сравнения можно рассмотреть следующие хорошо известные численные методы.

| Задача | Методы для сравнения |
|---|---|
| **Дифференцирование** | Правая разность, левая разность, центральная разность |
| **Интегрирование** | Формула прямоугольников (левых/правых/средних), формула трапеций, формула Симпсона |
| **Решение ОДУ** | Метод Эйлера (явный/неявный), метод Рунге-Кутты 2-го и 4-го порядка |

#### Краткое описание численных методов

##### Численное дифференцирование

1. **Левая разность (Left Difference)**
<br>
$$
f'(x_0) \approx \frac{f(x_0) - f(x_0 - h)}{h}
$$
<br>
Порядок точности: $ O(h)$.


2. **Правая разность (Right Difference)**
<br>
$$
f'(x_0) \approx \frac{f(x_0 + h) - f(x_0)}{h}
$$
<br>
Порядок точности: $O(h)$.

3. **Центральная разность (Central Difference)**
<br>
$$
f'(x_0) \approx \frac{f(x_0 + h) - f(x_0 - h)}{2h}
$$
<br>
Порядок точности: $ O(h^2) $.


##### Численное интегрирование (Составные квадратурные формулы)

1. **Составная формула левых прямоугольников (Left Rectangle)**
<br>
$$
\int_a^b f(x) \, dx \approx h \sum_{i=0}^{n-1} f(x_i), \quad h = \frac{b-a}{n}, \quad x_i = a + i \cdot h
$$
<br>
Порядок точности: $ O(h) $.



2. **Составная формула правых прямоугольников (Right Rectangle)**
<br>
$$
\int_a^b f(x) \, dx \approx h \sum_{i=1}^{n} f(x_i), \quad h = \frac{b-a}{n}, \quad x_i = a + i \cdot h
$$
<br>
Порядок точности: $ O(h) $.

3. **Составная формула средних прямоугольников (Midpoint Rectangle)**
<br>
$$
\int_a^b f(x) \, dx \approx h \sum_{i=0}^{n-1} f\left( \frac{x_i + x_{i+1}}{2} \right), \quad h = \frac{b-a}{n}
$$
<br>
Порядок точности: $ O(h^2) $.

4. **Составная формула трапеций (Trapezoidal Rule)**
<br>
$$
\int_a^b f(x) \, dx \approx h \left( \frac{f(x_0)}{2} + \sum_{i=1}^{n-1} f(x_i) + \frac{f(x_n)}{2} \right), \quad h = \frac{b-a}{n}
$$
<br>
Порядок точности: $ O(h^2) $.

5. **Формула Симпсона (Simpson's Rule)**
<br>
$$
\int_a^b f(x) \, dx \approx \frac{h}{3} \left( f(x_0) + 4\sum_{i \text{ нечет}} f(x_i) + 2\sum_{i \text{ чет}} f(x_i) + f(x_n) \right), \quad h = \frac{b-a}{n}, \quad n \text{ — чётное}
$$
<br>
Порядок точности: $ O(h^4) $


##### Численное решение ОДУ

1. **Явный метод Эйлера (Explicit Euler)**
<br>
Задача: $ y' = f(t, y), \quad y(t_0) = y_0 $
<br>
$$
y_{n+1} = y_n + h \cdot f(t_n, y_n)
$$
<br>
Порядок точности: $ O(h) $


2. **Неявный метод Эйлера (Implicit Euler)**
<br>
$$
y_{n+1} = y_n + h \cdot f(t_{n+1}, y_{n+1})
$$
<br>
Порядок точности: $ O(h) $.<br> 
**Замечание.** Требуется решать нелинейное уравнение на каждом шаге.

4. **Метод Рунге-Кутты 2-го порядка (RK2 / Heun's Method)**
<br>
$$
\begin{aligned}
k_1 &= f(t_n, y_n) \\
k_2 &= f(t_n + h, y_n + h \cdot k_1) \\
y_{n+1} &= y_n + \frac{h}{2} (k_1 + k_2)
\end{aligned}
$$<br>
Порядок точности: $ O(h^2) $.

5. **Метод Рунге-Кутты 4-го порядка (RK4)**
<br>
$$
\begin{aligned}
k_1 &= f(t_n, y_n) \\
k_2 &= f(t_n + \frac{h}{2}, y_n + \frac{h}{2} k_1) \\
k_3 &= f(t_n + \frac{h}{2}, y_n + \frac{h}{2} k_2) \\
k_4 &= f(t_n + h, y_n + h k_3) \\
y_{n+1} &= y_n + \frac{h}{6} (k_1 + 2k_2 + 2k_3 + k_4)
\end{aligned}
$$<br>
Порядок точности: $ O(h^4)$.

### Выбор шага

В данном примере рассматриваются сеточные методы.

Исходя из этого шаг $ h $ должен последовательно уменьшаться:

$$
h = 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001, \ldots
$$

Или по степеням 10:

$$
h = 10^{-1}, 10^{-2}, 10^{-3}, \ldots, 10^{-15}.
$$

### Критерии сравнения

| Критерий | Описание |
|---|---|
| **Фактическая погрешность** | $ \varphi(h) = \| \text{точное решение} - \text{приближенное решение} \| $ |
| **Порядок точности** | $ p = \lim_{h \to 0} \frac{\log \varphi(h)}{\log h} $ |
| **Константа** | $ C = \lim_{h \to 0} \frac{\varphi(h)}{h^p} $ |
| **Устойчивость** | Поведение погрешности при очень малых $ h $ |

## Сбор и обработка экспериментальных данных

### Форма записи результатов

| $ h $ | $ \varphi(h) $ | $ \lg h $ | $ \lg \varphi(h) $ | $ p_{\text{эксп}} $ |
|---|---|---|---|---|
| $ 10^{-1} $ | $ 0.1234 $ | -1.000 | -0.909 | — |
| $ 10^{-2} $ | $ 0.0123 $ | -2.000 | -1.910 | 1.001 |
| $ 10^{-3} $ | $ 0.0012 $ | -3.000 | -2.911 | 1.001 |
| ... | ... | ... | ... | ... |

### Вычисление экспериментального порядка точности

Для двух последовательных шагов $ h_1 $ и $ h_2 $:

$$
p_{\text{эксп}} = \frac{\log(\varphi(h_1) / \varphi(h_2))}{\log(h_1 / h_2)}.
$$

Если $ h_1 = 2h_2 $, то формула упрощается:

$$
p_{\text{эксп}} = \log_2 \left( \frac{\varphi(2h)}{\varphi(h)} \right).
$$

### Оценка константы $ C $

$$
C \approx \frac{\varphi(h)}{h^p}.
$$

Берется среднее по нескольким значениям $ h $.

## Визуализация результатов

### lg-lg графики

Строятся по парам $ (\lg h, \lg \varphi(h)) $.

**Свойства:**
- Если $ \varphi(h) \approx C \cdot h^p $, то график — прямая с угловым коэффициентом $ p $.
- Свободный член $ a = \lg C $.

### Зоны на lg-lg графике


| Зона | Описание |
|---|---|
| **Зона 1** | $ h $ велико, метод не работает, ошибка не следует закону $ C \cdot h^p $ |
| **Зона 2** | Асимптотическая зона, $ \varphi(h) \approx C \cdot h^p $, график — прямая |
| **Зона 3** | Ошибка округления доминирует, график "загибается" вверх |

### Оптимальный шаг

Точка минимума полной погрешности:

$$
\varphi_{\text{полн}}(h) = \varphi_{\text{аппрокс}}(h) + \varphi_{\text{округл}}(h).
$$

$ h_{\text{opt}} $ — шаг, при котором $ \varphi_{\text{полн}}(h) $ минимальна.

## Анализ устойчивости

### Влияние погрешностей округления

При очень малых $ h $ возникает катастрофическая отмена — вычитание близких чисел.

**Пример для правой разности:**

$$
\varphi_{\text{округл}} \approx \frac{2\varepsilon}{h} \cdot |f(x_0)|,
$$

где $ \varepsilon \approx 10^{-16} $ — машинный эпсилон.

```python
eps = 1.0
while 1.0 + eps != 1.0:
    eps /= 2
print(eps * 2)
```

### Полная погрешность

$$
\varphi_{\text{полн}}(h) \approx C \cdot h^p + \frac{D}{h}
$$

где $ D $ — константа, зависящая от $ f(x_0) $ и $ \varepsilon $.

### Оптимальный шаг (аналитическая оценка)

Из условия $ \frac{d}{dh} \varphi_{\text{полн}}(h) = 0 $:

$$
h_{\text{opt}} = \left( \frac{D}{p \cdot C} \right)^{1/(p+1)}
$$

## Сравнение методов

### Сравнение по порядку точности

| Метод | Порядок $ p $ | Комментарий |
|---|---|---|
| Левая/правая разность | 1 | Медленная сходимость |
| Центральная разность | 2 | Хороший выбор для гладких функций |
| Метод трапеций | 2 | Для интегрирования |
| Метод Симпсона | 4 | Очень точный для гладких функций |

### Сравнение по константе $ C $

При одинаковом $ p $ метод с меньшим $ C $ точнее.

**Пример** (интегрирование методом прямоугольников):

| Метод | $ C $ | Комментарий |
|---|---|---|
| Формула левых прямоугольников | $ 1/2 $ | Хуже |
| Формула правых прямоугольников | $ 1/2 $ | Хуже |
| Формула средних прямоугольников | $ 1/24 $ | Лучше |

### Сравнение по устойчивости

| Метод | Устойчивость при малых $ h $ |
|---|---|
| Левая/правая разность | Умеренная |
| Центральная разность | Лучше (ошибка округления $ O(1/h^2) $) |
| Метод Рунге-Кутты 4-го порядка | Хорошая для **нежестких** задач |
