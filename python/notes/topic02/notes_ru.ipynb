## Условные операторы (Управление потоком)

- **Конструкция `if`**
    - Синтаксис: отступы (индентация) как способ выделения блоков кода.
    - Операторы сравнения: `==`, `!=`, `>`, `<`, `>=`, `<=`.

- **Логические операторы**
    - `and`, `or`, `not`. Приоритет выполнения.
    - Проверка на вхождение (`in`).

- **Ветвления**
    - `if-else` (альтернатива).
    - `if-elif-else` (множественный выбор).

- **Тернарный оператор** (краткая запись):
    - `x = "Да" if age > 18 else "Нет"`.

- **Вложенные условия** (когда условие находится внутри другого условия).

## Циклы

- **Цикл `while`** (с предусловием)
    - Риск бесконечного цикла и как его избежать (`break`).

```python
i = 0
while i < 5:
    print(f"Итерация {i}")
    i += 1  # увеличиваем счетчик

# Вывод:
# Итерация 0
# Итерация 1
# Итерация 2
# Итерация 3
# Итерация 4
```

- **Цикл `for`** (итератор)
    - Обход строк, списков, кортежей.
    - Функция `range()`: генерация последовательностей чисел.
    - Обход словарей (по ключам, по значениям, по парам).

```python
for i in range(5):
    print(i)
```

```python
for _ in range(3):
    print("Привет!")
```

- **Управляющие операторы в циклах**
    - `break` – экстренный выход из цикла.
    - `continue` – переход к следующей итерации (пропуск).
    - `else` у цикла (выполняется, если не было `break`).

- **Генераторы списков** (List Comprehensions) как компактная альтернатива циклам.
    - *Пример:* `[x**2 for x in range(10) if x % 2 == 0]`

## Функции (Абстракция и Переиспользование)

- **Определение функции (`def`) и вызов**

- **Аргументы и параметры**
    - Позиционные и именованные (keyword) аргументы.
    - Аргументы по умолчанию (`def func(x=10)`).
    - `*args` и `**kwargs` (переменное число аргументов).

- **Возврат значения (`return`)**
    - Отличие `return` от `print()`.
    - Возврат нескольких значений (как кортеж).

- **Области видимости переменных**
    - Локальные (`local`) и Глобальные (`global`).
    - Ключевое слово `global` (почему его лучше не использовать).

- **Анонимные функции (`lambda`)**
    - Использование вместе с `map()`, `filter()`, `sorted()`.

```python
# Простая lambda-функция
square = lambda x: x ** 2

# Использование
print(square(5))  # 25

# Аналог с обычной функцией
def square_def(x):
    return x ** 2
```

```python
# map() - применяет функцию к каждому элементу
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, numbers))
print(squared)  # [1, 4, 9, 16, 25]

# filter() - фильтрует элементы по условию
numbers = [1, 2, 3, 4, 5, 6]
even = list(filter(lambda x: x % 2 == 0, numbers))
print(even)  # [2, 4, 6]

# sorted() - сортирует с ключом
people = [('Alice', 25), ('Bob', 20), ('Charlie', 30)]
sorted_by_age = sorted(people, key=lambda x: x[1])
print(sorted_by_age)  # [('Bob', 20), ('Alice', 25), ('Charlie', 30)]
```

## Классы и Объекты (Основы ООП)

- **Парадигма ООП** (зачем это нужно: моделирование реального мира).

- **Синтаксис класса**
    - Ключевое слово `class`.
    - Конструктор `__init__` (инициализация атрибутов объекта).
    - `self` – ссылка на сам объект.

- **Атрибуты и Методы**
    - Атрибуты экземпляра (в `__init__`).
    - Методы экземпляра (обычные функции внутри класса).

- **Магические методы (кратко)**
    - `__str__` (строковое представление для пользователей) и `__repr__` (для разработчиков).

- **Инкапсуляция (кратко)**
    - Понятие публичных (`public`) и защищенных (`_`) атрибутов.
    - `@property` (геттеры/сеттеры) — если позволяет время.
