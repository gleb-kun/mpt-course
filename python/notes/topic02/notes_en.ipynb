## Conditional Statements (Flow Control)

- **The 'if' Construct**
    - Syntax: Indentation as a way to delimit blocks of code.
    - Comparison Operators: `==`, `!=`, `>`, `<`, `>=`, `<=`.

- **Logical Operators**
    - `and`, `or`, `not`. Execution Precedence.
    - Inclusion Test (`in`).

- **Branching**
    - `if-else` (alternative).
    - `if-elif-else` (multiple choice).

- **Ternary Operator** (shortcut):
    - `x = "Yes" if age > 18 else "No"`.

- **Nested Conditionals** (when a condition is contained within another condition).

## Loops

- **`while`** loop (with precondition)
    - The risk of an infinite loop and how to avoid it (`break`).

```python
i = 0
while i < 5:
    print(f"Iteration {i}")
    i += 1   # increment the counter

# Output:
# Iteration 0
# Iteration 1
# Iteration 2
# Iteration 3
# Iteration 4
```

- **`for` loop** (iterator)
    - Iterating over strings, lists, and tuples.
    - `range()` function: generating sequences of numbers.
    - Iterating over dictionaries (by keys, by values, by pairs).

```python
for i in range(5):
    print(i)
```

```python
for _ in range(3):
    print("Hi!")
```

- **Control statements in loops**
    - `break` – emergency exit from a loop.
    - `continue` – skip to the next iteration.
    - `else` in a loop (executed if there was no `break`).

- **List comprehensions** as a compact alternative to loops.
    - *Example:* `[x**2 for x in range(10) if x % 2 == 0]`

## Functions (Abstraction and Reuse)

- **Function Definition (`def`) and Call**

- **Arguments and Parameters**
    - Positional and Keyword Arguments.
    - Default Arguments (`def func(x=10)`).
    - `*args` and `**kwargs` (variable number of arguments).

- **Returning a Value (`return`)**
    - The Difference Between `return` and `print()`.
    - Returning Multiple Values ​​(as a Tuple).

- **Variable Scopes**
    - Local (`local`) and Global (`global`).
    - The `global` Keyword (Why It's Best Not to Use It).

- **Anonymous Functions (`lambda`)**
    - Using with `map()`, `filter()`, and `sorted()`.

```python
# Simple lambda function
square = lambda x: x ** 2

# Usage
print(square(5)) # 25

# Analog with a regular function
def square_def(x):
    return x ** 2
```

```python
# map() - applies a function to each element
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, numbers))
print(squared) # [1, 4, 9, 16, 25]

# filter() - filters elements by a condition
numbers = [1, 2, 3, 4, 5, 6]
even = list(filter(lambda x: x % 2 == 0, numbers))
print(even) # [2, 4, 6]

# sorted() - sorts by key
people = [('Alice', 25), ('Bob', 20), ('Charlie', 30)]
sorted_by_age = sorted(people, key=lambda x: x[1])
print(sorted_by_age) # [('Bob', 20), ('Alice', 25), ('Charlie', 30)]
```
