### Задание 1. Линейная алгебра. Метод Гаусса

#### Общая информация

Рассмотрим процедуру метода Гаусса в общем виде. Дана система линейных алгебраических уравнений (СЛАУ) порядка~$n$

$$
\begin{gathered}
    \begin{cases}
        a_{11}^{\left(0\right)}x_1+a_{12}^{\left(0\right)}x_2+\ldots+a_{1n}^{\left(0\right)}x_n=b_{1}^{\left(0\right)}\\
        a_{21}^{\left(0\right)}x_1+a_{22}^{\left(0\right)}x_2+\ldots+a_{2n}^{\left(0\right)}x_n=b_{2}^{\left(0\right)}\\
        \ldots\\
        a_{n1}^{\left(0\right)}x_1+a_{n2}^{\left(0\right)}x_2+\ldots+a_{nn}^{\left(0\right)}x_n=b_{n}^{\left(0\right)}
    \end{cases}.
\end{gathered}\tag{1}
$$

Предположим, что коэффициент $a_{11}^{\left(0\right)}$, называемый *ведущим элементом* первой строки, не равен нулю.

Найдем величины
$$
\mu_{i1} = \frac{a^{\left(0\right)}_{i1}}{a^{\left(0\right)}_{11}},\quad i=2,3,\ldots, n.\tag{2}
$$
    
Далее исключаем неизвестную $x_1$ из каждого уравнения системы, начиная со второго. Для этого вычитаем из второго, третьего$,\ldots, n$-го уравнения системы (1) первое уравнение умноженное на коэффициент $\mu_{21}$, $\mu_{31},\ldots, \mu_{n1}$ соответственно. Преобразованная система уравнений имеет вид

$$
    	\begin{cases}
    		a_{11}^{\left(0\right)}x_1+a_{12}^{\left(0\right)}x_2+a_{13}^{\left(0\right)}x_3+\ldots+a_{1n}^{\left(0\right)}x_n=b_1^{\left(0\right)}\\
    		\hspace{15.3mm} a_{22}^{\left(1\right)}x_2+a_{23}^{\left(1\right)}x_3+\ldots+a_{2n}^{\left(1\right)}x_n=b_{2}^{\left(1\right)}\\
    		\hspace{15.3mm}\ldots\\
    		\hspace{15.3mm} a_{n2}^{\left(1\right)}x_2+a_{n3}^{\left(1\right)}x_3+\ldots+a_{nn}^{\left(1\right)}x_n=b_{n}^{\left(1\right)}\\
    	\end{cases},
    	\tag{3}
$$
где
$$
a_{ij}^{\left(1\right)}=a_{ij}^{\left(0\right)}-\mu_{i1}\cdot a^{\left(0\right)}_{1j},\quad b_{i}^{\left(1\right)}=b_{i}^{\left(0\right)}-\mu_{i1}{b^{\left(0\right)}_{1}},\quad i,j=2,3,\ldots,n.\tag{4}
$$
Поступаем аналогично со следующим уравнением из преобразованной системы. В конечном итоге приводим исходную систему к эквивалентной системе
$$
    	\begin{cases}
    		a_{11}^{\left(0\right)}x_1+a_{12}^{\left(0\right)}x_2+a_{13}^{\left(0\right)}x_3+\ldots+a_{1n}^{\left(0\right)}x_n=b_1^{\left(0\right)}\\
    		\hspace{15.3mm} a_{22}^{\left(1\right)}x_2+a_{23}^{\left(1\right)}x_3+\ldots+a_{2n}^{\left(1\right)}x_n=b_{2}^{\left(1\right)}\\
    		\hspace{15.3mm}\ldots\\
    		\hspace{47.5mm} a_{nn}^{\left(n-1\right)}x_n=b_{n}^{\left(n-1\right)}\\
    	\end{cases}
    	\tag{5}
$$
с верхней треугольной матрицей $\textbf{A}^{\left(n-1\right)}$.
    
Далее обратным ходом из системы (5) последовательно находим неизвестные $x_n$, $x_{n-1}$, $\ldots$,~$x_1$.

#### Число арифметических операций в методе Гаусса

Узнаем количество арифметических операций, выполняемых при решении СЛАУ порядка $n$.

Вычисления 1-го шага прямого хода с использованием формул (2) и (5) требуют выполнения $n-1$ делений, $\left(n-1\right)n$ умножений и $\left(n-1\right)n$ вычитаний, таким образом общее число арифметических операций, выполняемых на первом шаге, равно
$$
        Q_1=2\left(n-1\right)^2+3\left(n-1\right).
$$
    Аналогично, на втором шаге требуется операций
$$
        Q_2=2\left(n-2\right)^2+3\left(n-2\right).
$$
Не трудно догадаться, что для исключения неизвестной $x_i$ надо затратить операций
$$
    Q_i=2\left(n-i\right)^2+3\left(n-i\right).
$$

Подсчитаем общее число необходимых арифметических операций в прямом ходе
$$
        \begin{gathered}
            Q_{\text{forward}}=\sum^{n-1}_{i=1}Q_i=2\sum_{i=1}^{n-1}\left(n-i\right)^2+3\sum_{i=1}^{n-1}\left(n-i\right)=2\sum_{i=1}^{n-1}i^2+3\sum_{i=1}^{n-1}i=\\=\frac{2\left(n-1\right)n\left(2n-1\right)}{6}+\frac{3\left(n-1\right)n}{2}=\frac{2}{3}n^3+\frac{1}{2}n^2-\frac{7}{6}n.
        \end{gathered}
$$
Число необходимых арифметических действий в обратном ходе
$$
        Q_{\text{backward}}=n^2,
$$

Таким образом общее количество арифметических операций при решении СЛАУ методом Гаусса будет
$$
        Q=Q_{\text{forward}}+Q_{\text{backward}}=\frac{2}{3}n^3+\frac{3}{2}n^2-\frac{7}{6}n\approx\frac{2}{3}n^3.
$$

#### Вычисление определителей

Определитель матрицы при делении ее строки на ведущий элемент также делится на этот элемент, а при вычитании строки, умноженной на любое число, как известно, определитель новой матрицы равен определителю старой.
Если каждую строку треугольной матрицы $\textbf{A}^{\left(n-1\right)}$, полученной в ходе прямого хода метода Гаусса, разделить на соответствующий ведущий элемент $a_{{ii}}^{\left(i-1\right)}$, то ее определитель равен 1. Он получен в результате деления определителя $\det\left(\textbf{A}\right)$ исходной матрицы $\textbf{A}=\textbf{A}^{\left(0\right)}$ системы (1) на ведущие элементы на каждом шаге: $a_{11}^{\left(0\right)}$, $a_{22}^{\left(1\right)}$, $\ldots$, $a_{nn}^{\left(n-1\right)}$. Следовательно, если ни один из ведущих элементов не равен нулю, то
$$
        \det\left(\textbf{A}\right)
        =a_{11}^{\left(0\right)}\cdot a_{22}^{\left(1\right)}\cdot\ldots\cdot a_{nn}^{\left(n-1\right)}.
$$

Таким образом попутно при решении системы методом Гаусса можно вычислить определитель матрицы системы линейных алгебраических уравнений.

#### Выбор главного элемента

Во избежание деления на малый ведущий элемент рекомендуется осуществлять выбор наибольшего по модулю элемента и считать его ведущим.

#### Обращение матриц

Для нахождения матрицы $\textbf{A}^{-1}$ формально требуется решить $n$ систем, векторы правых частей которых представляют собой единичные орты. Однако, чтобы не осуществлять прямой ход $n$ раз, можно все преобразования делать в расширенной матрице размера $n \times 2n$ вида $\left(\textbf{A},\textbf{E}\right)$, просто расширив циклы по $j$ до значения $2n$ включительно.

## Геометрия

Для решения задач по геометрии воспользуемся библиотекой SymPy. В SymPy есть модуль `sympy.geometry`.

### Обзор основных функций модуля `sympy.geometry`.

Основные функции модуля `sympy.geometry` можно условно разделить на несколько групп:


#### 1. Создание новых геометрических объектов

Модуль предоставляет классы для создания геометрических объектов:

| Класс | Описание |
|-------|----------|
| `Point` | Точка |
| `Line` | Прямая |
| `Segment` | Отрезок |
| `Ray` | Луч |
| `Circle` | Окружность |
| `Ellipse` | Эллипс |
| `Triangle` | Треугольник |
| `Polygon` | Многоугольник |
| `Plane` | Плоскость (в 3D-пространстве) |

##### Производные фигуры

Наряду с базовыми объектами, доступны и производные фигуры, определяемые через уже существующие объекты:

- **Triangle** имеет методы для построения:
  - высот (`altitude`)
  - медиан (`median`)
  - биссектрис (`bisector`)
  - вписанной окружности (`incircle`)
  - описанной окружности (`circumcircle`)
  - прямой Эйлера (`EulerLine`)
  - точки Жергонна (`GergonnePoint`)
  - точки Нагеля (`NagelPoint`)

- **Circle** позволяет строить касательные (`tangent_lines`)

- **Дополнительные функции:**
  - построение серединного перпендикуляра (`perpendicular_bisector`)
  - построение параллельных и перпендикулярных прямых

##### Гибкость создания объектов

Один и тот же объект можно задать несколькими способами. Например, треугольник (`Triangle`) можно создать:

- по координатам трех точек
- по трем отрезкам (сторонам)
- по длинам сторон (в этом случае треугольник задается без привязки к координатной плоскости, и SymPy автоматически размещает его на плоскости)

**Важно:** Все конструкторы используют ключевые аргументы для указания параметров объектов, а не имя создаваемого объекта в качестве первого параметра (в отличие, например, от Maple).

#### 2. Функции, возвращающие характеристики объектов

Для каждого объекта доступны методы и функции, возвращающие различные геометрические свойства:

| Метод/Функция | Описание |
|---------------|----------|
| `coordinates` | Координаты |
| `area` | Площадь |
| `perimeter` | Периметр |
| `center` | Центр |
| `radius` | Радиус |
| `angle_between` | Угол между прямыми |
| `distance` | Расстояние между объектами |
| `equation` | Уравнение объекта в аналитическом виде |
| `type` | Тип объекта |
| `method` | Метод создания (для объектов, которые можно построить разными способами) |

##### Получение информации об объекте

- `print(object)` — для получения наиболее полной информации
- `object.__dict__` — для просмотра всех атрибутов
- `dir(object)` — для справки по доступным методам

#### 3. Функции проверки инцидентности и взаимного расположения

В модуле реализованы предикаты и методы для проверки геометрических свойств и взаимного расположения объектов:

| Функция/Метод | Описание |
|---------------|----------|
| `are_collinear` | Принадлежность трех или более точек одной прямой |
| `are_concyclic` | Принадлежность четырех или более точек одной окружности |
| `is_right` | Проверка, является ли треугольник прямоугольным |
| `is_equilateral` | Проверка, является ли треугольник равносторонним |
| `is_scalene` | Проверка, является ли треугольник разносторонним |
| `is_similar` | Проверка подобия треугольников |
| `intersect` | Проверка наличия пересечения и нахождение точек пересечения объектов |
| `contains` | Проверка принадлежности точки объекту |
| `is_convex` | Проверка выпуклости многоугольника |
| `is_parallel` | Проверка параллельности прямых |
| `is_perpendicular` | Проверка перпендикулярности прямых |

#### 4. Геометрические преобразования

Модуль поддерживает основные преобразования геометрических объектов:

| Метод | Описание |
|-------|----------|
| `translate` | Параллельный перенос |
| `rotate` | Поворот относительно точки |
| `reflect` | Симметрия (отражение) относительно точки или прямой |
| `scale` | Масштабирование (аналог гомотетии) |
| `transform` | Произвольные аффинные преобразования |

**Важно:** Эти методы применяются непосредственно к объектам и возвращают новые преобразованные объекты.

#### 5. Визуализация

Для отображения геометрических объектов используется:

- Модуль `sympy.plotting` (в частности, `Plot`)
- Интеграция с внешними библиотеками:
  - **Matplotlib**:
    - `plot_implicit` — для неявных кривых
    - `plot` — для параметрических кривых

**Примечание:** Непосредственного аналога команды `draw` с гибкой локальной/глобальной настройкой параметров отображения в SymPy нет. Однако с помощью Matplotlib можно настроить отображение каждого объекта индивидуально (цвет, толщина линии, стиль и т.д.).
