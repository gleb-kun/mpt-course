### Task 1. Linear Algebra. Gaussian Method

#### General Information

Let's look at the Gaussian method in general. A system of linear algebraic equations (SLAE) of order ~$n$ is given.

$$
\begin{gathered}
\begin{cases}
a_{11}^{\left(0\right)}x_1+a_{12}^{\left(0\right)}x_2+\ldots+a_{1n}^{\left(0\right)}x_n=b_{1}^{\left(0\right)}\\
a_{21}^{\left(0\right)}x_1+a_{22}^{\left(0\right)}x_2+\ldots+a_{2n}^{\left(0\right)}x_n=b_{2}^{\left(0\right)}\\
\ldots\\
a_{n1}^{\left(0\right)}x_1+a_{n2}^{\left(0\right)}x_2+\ldots+a_{nn}^{\left(0\right)}x_n=b_{n}^{\left(0\right)}
\end{cases}.
\end{gathered}\tag{1}
$$

Suppose that the coefficient $a_{11}^{\left(0\right)}$, called the *pivot* of the first row, is nonzero.

Let's find the values
$$
\mu_{i1} = \frac{a^{\left(0\right)}_{i1}}{a^{\left(0\right)}_{11}},\quad i=2,3,\ldots, n.\tag{2}
$$

Next, we eliminate the unknown $x_1$ from each equation of the system, starting with the second. To do this, we subtract the first equation multiplied by the coefficients $\mu_{21}$, $\mu_{31},\ldots, and \mu_{n1}$ from the second, third, and nth equations of system (1), respectively. The transformed system of equations has the form

$$ 
\begin{cases} 
a_{11}^{\left(0\right)}x_1+a_{12}^{\left(0\right)}x_2+a_{13}^{\left(0\right)}x_3+\ldots+a_{1n}^{\left(0\right)}x_n=b_1^{\left(0\right)}\\ 
\hspace{15.3mm} a_{22}^{\left(1\right)}x_2+a_{23}^{\left(1\right)}x_3+\ldots+a_{2n}^{\left(1\right)}x_n=b_{2}^{\left(1\right)}\\ 
\hspace{15.3mm}\ldots\\ 
\hspace{15.3mm} a_{n2}^{\left(1\right)}x_2+a_{n3}^{\left(1\right)}x_3+\ldots+a_{nn}^{\left(1\right)}x_n=b_{n}^{\left(1\right)}\\ 
\end{cases}, 
\tag{3}
$$
Where
$$
a_{ij}^{\left(1\right)}=a_{ij}^{\left(0\right)}-\mu_{i1}\cdot a^{\left(0\right)}_{1j},\quad b_{i}^{\left(1\right)}=b_{i}^{\left(0\right)}-\mu_{i1}{b^{\left(0\right)}_{1}},\quad i,j=2,3,\ldots,n.\tag{4}
$$
We proceed similarly with the next equation from the transformed system. Finally, we reduce the original system to an equivalent system
$$
\begin{cases}
a_{11}^{\left(0\right)}x_1+a_{12}^{\left(0\right)}x_2+a_{13}^{\left(0\right)}x_3+\ldots+a_{1n}^{\left(0\right)}x_n=b_1^{\left(0\right)}\\
\hspace{15.3mm} a_{22}^{\left(1\right)}x_2+a_{23}^{\left(1\right)}x_3+\ldots+a_{2n}^{\left(1\right)}x_n=b_{2}^{\left(1\right)}\\
\hspace{15.3mm}\ldots\\
\hspace{47.5mm} a_{nn}^{\left(n-1\right)}x_n=b_{n}^{\left(n-1\right)}\\
\end{cases}
\tag{5}
$$
with an upper triangular matrix $\textbf{A}^{\left(n-1\right)}$.

Next, using the backward step from system (5), we sequentially find the unknowns $x_n$, $x_{n-1}$, $\ldots$,~$x_1$.

#### Number of arithmetic operations in the Gaussian method

Let's find the number of arithmetic operations performed when solving a SLAE of order $n$.

Calculating the first step of the forward approach using formulas (2) and (5) requires $n-1$ divisions, $\left(n-1\right)n$ multiplications, and $\left(n-1\right)n$ subtractions. Thus, the total number of arithmetic operations performed in the first step is
$$
Q_1=2\left(n-1\right)^2+3\left(n-1\right).
$$
Similarly, the second step requires operations
$$
Q_2=2\left(n-2\right)^2+3\left(n-2\right).
$$
It's not hard to guess that eliminating the unknown $x_i$ requires operations
$$
Q_i=2\left(n-i\right)^2+3\left(n-i\right).
$$

Let's calculate the total number of required arithmetic operations in the forward step
$$
\begin{gathered} 
Q_{\text{forward}}=\sum^{n-1}_{i=1}Q_i=2\sum_{i=1}^{n-1}\left(n-i\right)^2+3\sum_{i=1}^{n-1}\left(n-i\right)=2\sum_{i=1}^{n-1}i^2+3\sum _{i=1}^{n-1}i=\\=\frac{2\left(n-1\right)n\left(2n-1\right)}{6}+\frac{3\left(n-1\right)n}{2}=\frac{2}{3}n^3+\frac{1}{2}n^2-\frac{7}{6}n. 
\end{gathered}
$$
Number of required arithmetic operations in the backward direction
$$
Q_{\text{backward}}=n^2,
$$

Thus, the total number of arithmetic operations when solving a SLAE using the Gauss method will be
$$
Q=Q_{\text{forward}}+Q_{\text{backward}}=\frac{2}{3}n^3+\frac{3}{2}n^2-\frac{7}{6}n\approx\frac{2}{3}n^3.
$$

#### Calculating Determinants

When dividing a matrix row by its leading element, the determinant is also divided by that element. When subtracting a row multiplied by any number, the determinant of the new matrix is ​​known to be equal to the determinant of the old one.
If each row of the triangular matrix $\textbf{A}^{\left(n-1\right)}$ obtained by the forward step of the Gaussian method is divided by the corresponding leading element $a_{{ii}}^{\left(i-1\right)}$, then its determinant is equal to 1. It is obtained by dividing the determinant $\det\left(\textbf{A}\right)$ of the original matrix $\textbf{A}=\textbf{A}^{\left(0\right)}$ of system (1) by the leading elements at each step: $a_{11}^{\left(0\right)}$, $a_{22}^{\left(1\right)}$, $\ldots$, $a_{nn}^{\left(n-1\right)}$. Therefore, if none of the leading elements is zero, then
$$
\det\left(\textbf{A}\right)
=a_{11}^{\left(0\right)}\cdot a_{22}^{\left(1\right)}\cdot\ldots\cdot a_{nn}^{\left(n-1\right)}.
$$

Thus, while solving the system using the Gauss method, one can simultaneously calculate the determinant of the matrix of the system of linear algebraic equations.

#### Selecting the Leading Element

To avoid dividing by a small leading element, it is recommended to select the element with the largest absolute value and consider it the leading element.

#### Matrix Inversion

To find the matrix $\textbf{A}^{-1}$, one formally needs to solve $n$ systems whose right-hand side vectors are unit unit vectors. However, to avoid performing the forward step $n$ times, all transformations can be performed in an extended matrix of size $n \times 2n$ of the form $\left(\textbf{A},\textbf{E}\right)$, simply expanding the cycles in $j$ up to and including the value $2n$.

## Geometry

To solve geometry problems, we'll use the SymPy library. SymPy includes the `sympy.geometry` module.

### Overview of the main functions of the `sympy.geometry` module.

The main functions of the `sympy.geometry` module can be divided into several groups:

#### 1. Creating new geometric objects

The module provides classes for creating geometric objects:

| Class | Description |
|-------|---------|
| `Point` | Point |
| `Line` | Line |
| `Segment` | Segment |
| `Ray` | Ray |
| `Circle` | Circle |
| `Ellipse` | Ellipse |
| `Triangle` | Triangle |
| `Polygon` | Polygon |
| `Plane` | Plane (in 3D space) |

##### Derived Shapes

In addition to basic objects, derived shapes defined using existing objects are also available:

- **Triangle** has methods for constructing:
- altitudes (`altitude`)
- medians (`median`)
- bisectors (`bisector`)
- incircles (`incircle`)
- circumcircles (`circumcircle`)
- Euler lines (`EulerLine`)
- Gergonne points (`GergonnePoint`)
- Nagel points (`NagelPoint`)

- **Circle** allows you to construct tangents (`tangent_lines`)

- **Additional functions:**
- constructing the perpendicular bisector (`perpendicular_bisector`)
- constructing parallel and perpendicular lines

##### Flexibility of Object Creation

The same object can be defined in multiple ways. For example, a triangle (`Triangle`) can be created:

- by the coordinates of three points
- by three line segments (sides)
- by the lengths of the sides (in this case, the triangle is defined without reference to the coordinate plane, and SymPy automatically places it on the plane)

**Important:** All constructors use keyword arguments to specify object parameters, rather than the name of the object being created as the first parameter (unlike, for example, Maple).

#### 2. Functions Returning Object Characteristics

Each object has methods and functions that return various geometric properties:

| Method/Function | Description |
|--------------|----------|
| `coordinates` | Coordinates |
| `area` | Area |
| `perimeter` | Perimeter |
| `center` | Center |
| `radius` | Radius |
| `angle_between` | Angle between lines |
| `distance` | Distance between objects |
| `equation` | Object equation in analytical form |
| `type` | Object type |
| `method` | Creation method (for objects that can be constructed in different ways) |

##### Getting information about an object

- `print(object)` — for more detailed information
- `object.__dict__` — to view all attributes
- `dir(object)` — for help with available methods

#### 3. Functions for checking incidence and relative position

The module implements predicates and methods for checking the geometric properties and relative position of objects:

| Function/Method | Description |
|--------------|----------|
| `are_collinear` | Whether three or more points belong to the same line |
| `are_concyclic` | Four or more points belong to the same circle |
| `is_right` | Check if a triangle is right-angled |
| `is_equilateral` | Check if a triangle is equilateral |
| `is_scalene` | Check if a triangle is scalene |
| `is_similar` | Check if triangles are similar |
| `intersect` | Check if objects intersect and find their intersection points |
| `contains` | Check if a point belongs to an object |
| `is_convex` | Check if a polygon is convex |
| `is_parallel` | Check if lines are parallel |
| `is_perpendicular` | Check if lines are perpendicular |

#### 4. Geometric Transformations

The module supports basic transformations of geometric objects:

| Method | Description |
|-------|----------|
| `translate` | Parallel translation |
| `rotate` | Rotation about a point |
| `reflect` | Symmetry (reflection) about a point or line |
| `scale` | Scaling (analogous to homothety) |
| `transform` | Arbitrary affine transformations |

**Important:** These methods are applied directly to objects and return new transformed objects.

#### 5. Visualization

The following is used to display geometric objects:

- The `sympy.plotting` module (in particular, `Plot`)
- Integration with external libraries:
- **Matplotlib**:
- `plot_implicit` — for implicit curves
- `plot` — for parametric curves

**Note:** SymPy does not have a direct equivalent of the `draw` command with flexible local/global display settings. However, with Matplotlib you can customize the display of each object individually (color, line thickness, style, etc.).
