## Python native tools

### venv (Built-in module)

1. **Step 1.** Create a virtual environment

```bash
# Create venv in the .venv folder (default name)
python -m venv .venv

# Or with a specific Python version
python3.11 -m venv .venv
```

2. **Step 2.** Activate venv

Windows:
```bash
.venv\Scripts\activate
```

macOS/Linux:
```
source .venv/bin/activate
```

3. **Step 3.** Checking Activation

```bash
# (.venv) will appear before the prompt
which python # Linux/macOS - will show the path inside venv
where python # Windows
```

4. **Step 4.** Package Management

Installing packages:

```bash
pip install requests
pip install django flask
```

Creating requirements.txt:

```bash
# Export all installed packages
pip freeze > requirements.txt
```

Installing from requirements.txt:
```bash
pip install -r requirements.txt
```

5. Deactivation

```bash
deactivate
```

6. Removing venv

```bash
# Just delete the folder
rm -rf .venv # Linux/macOS
rmdir /s .venv # Windows
```

Dependencies.txt file (example)

```txt
# Usually called requirements.txt
django==4.2.7
requests==2.31.0
python-dotenv>=1.0.0
pytest~=7.4.0
```

Version specifiers:

- `==` — exact version

- `>=` — not lower than

- `~=` — compatible version (`7.4.0` ≤ version < `8.0.0`)

Useful Commands

```bash
# Viewing Installed Packages
pip list

# Checking Outdated Packages
pip list --outdated

# Upgrading Pip
python -m pip install --upgrade pip
```

## Conda: An Ecosystem, Not Just an Environment

### 0. Installing `Conda`
Conda is often not available in official repositories. For example, in Arch, Conda is only available in the AUR repository.
Therefore, it's best to download Conda manually from the following link: https://repo.anaconda.com/archive/. For example:

```bash
curl -O https://repo.anaconda.com/archive/Anaconda3-2024.06-1-Linux-x86_64.sh
bash Anaconda3-2024.06-1-Linux-x86_64.sh
```

Next, run the `Anaconda3-2024.06-1-Linux-x86_64.sh` script; it will perform the installation process.

### 1. Creating an Environment

```bash
# Create an environment with Python 3.11
conda create -n llmsim-env python=3.11

# Specifying specific packages
conda create -n llmsim-env python=3.11 numpy pandas

# Specifying a channel (e.g., conda-forge)
conda create -n llmsim-env -c conda-forge python=3.11
```

### 2. Activating the Environment

```bash
conda activate llmsim-env
```

### 3. Using the Environment as a Kernel for JupyterLab:

Create and activate your Conda environment (if you haven't already):
```bash
conda create -n environment_name python=3.11
conda activate environment_name
```

Install ipykernel in this environment:

```bash
conda install ipykernel
```

Or via pip, if you prefer: `pip install ipykernel`.

Register the environment as a kernel for Jupyter:

```bash
python -m ipykernel install --user --name=environment_name --display-name="The name you will see in Jupyter"
```

- `--name`: This is the internal technical name of the kernel.
- `--display-name`: This is the name that will be displayed in the JupyterLab launcher.

After that, refresh the JupyterLab page in your browser – your new kernel should appear in the list of available kernels.

You must first install the package into the environment.

After installation, you must add the environment (kernel) to JupyterLab. The environment name used here is llmsim-env:

### 4. `environment.yml` file:

```yml
name: my_project
channels:
    - conda-forge  # The most popular channel with new versions
    - defaults
dependencies:
    - python=3.11
    - numpy=1.24
    - pandas
    - pip          # Be sure to add pip if you plan to use it
    - pip:
        - requests     # This will be installed via pip, as conda-forge may not have the latest version
```

Usage:

```bash
conda env create -f environment.yml
```
