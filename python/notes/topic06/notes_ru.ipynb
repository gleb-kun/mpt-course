#### Двумерный график (`plt.plot`)

Команда `plt.plot` — строит двумерные графики. Примеры использования функции `plt.plot`:

- 1. Явно заданных функций $y=f(x)$.
```python
plt.plot(x, y, опции)
```

- 2. Параметрически заданных кривых $x=f(t)$, $y=g(t)$.

```python
t = np.linspace(0, 2*np.pi, 100)
x = np.cos(t)
y = np.sin(t)
plt.plot(x, y, опции)
```

- 3. Множества точек.

```python
# Способ 1 — список пар координат
plt.plot([[x1, x2, x3], [y1, y2, y3]], 'o', опции)

# Способ 2 — отдельно массивы X и Y
plt.plot([x1, x2, x3], [y1, y2, y3], 'o', опции)
```

- 4. Несколько кривых на одном графике.
```python
plt.plot(x1, y1, опции1, x2, y2, опции2, ...)
```

##### Основные опции (задаются как именованные аргументы)

| Опция | Значение (по умолчанию подчёркнуто) | Пояснение |
|-------|--------------------------------------|-----------|
| `color` / `c` | `'b'` (синий), `'r'`, `'g'`, `'k'` (чёрный), `'m'`, `'c'`, `'y'`, `'w'` | Цвет линии/маркеров |
| `linestyle` / `ls` | `'-'` (сплошная), `'--'` (штрих), `'-.'` (штрих-пунктир), `':'` (пунктир), `''` (без линии) | Стиль линии |
| `marker` | `''` (нет), `'o'`, `'s'`, `'^'`, `'D'`, `'*'`, `'+'`, `'x'`, `'.'` | Тип маркера для точек |
| `linewidth` / `lw` | `1.0` | Толщина линии в пунктах |
| `markersize` / `ms` | `6.0` | Размер маркера |
| `alpha` | `1.0` (непрозрачный) | Прозрачность (0–1) |
| `label` | `''` | Подпись для легенды |

````python
# Генерация данных
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x + 0.5) * 0.8
y4 = np.cos(x + 0.5) * 0.8

# Создание фигуры и осей
plt.figure(figsize=(10, 6))

# Пример 1: синяя сплошная линия с круглыми маркерами
plt.plot(x, y1, 
         color='b',           # синий цвет
         linestyle='-',       # сплошная линия
         marker='o',          # круглые маркеры
         linewidth=2.0,       # толщина линии 2 пункта
         markersize=8.0,      # размер маркеров 8
         alpha=0.8,           # прозрачность 80%
         label='sin(x)')

# Пример 2: красная штриховая линия с квадратными маркерами
plt.plot(x, y2, 
         color='r',           # красный цвет
         linestyle='--',      # штриховая линия
         marker='s',          # квадратные маркеры
         linewidth=1.5,       
         markersize=6.0,
         alpha=0.9,
         label='cos(x)')

# Пример 3: зелёная штрих-пунктирная линия с треугольными маркерами
plt.plot(x, y3, 
         color='g',           
         linestyle='-.',      # штрих-пунктир
         marker='^',          # треугольные маркеры
         linewidth=2.5,
         markersize=9.0,
         alpha=0.7,
         label='sin(x+0.5)*0.8')

# Пример 4: пурпурная пунктирная линия с ромбовидными маркерами
plt.plot(x, y4, 
         color='m',           
         linestyle=':',       # пунктирная линия
         marker='D',          # ромбовидные маркеры
         linewidth=2.0,
         markersize=7.0,
         alpha=0.6,
         label='cos(x+0.5)*0.8')

# Добавление заголовка и подписей осей
plt.title('Пример использования основных опций matplotlib', fontsize=14)
plt.xlabel('X', fontsize=12)
plt.ylabel('Y', fontsize=12)

# Добавление сетки
plt.grid(True, alpha=0.3)

# Добавление легенды
plt.legend(loc='upper right', fontsize=10)

# Отображение графика
plt.tight_layout()
plt.show()
````

#### Трехмерный график (`plot3d`)

Для 3D-графиков используется модуль `mpl_toolkits.mplot3d`.

##### Явная функция $z=f(x,y)$

````python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax.plot_surface(X, Y, Z)   # поверхность
plt.show()
````

````python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax.plot_wireframe(X, Y, Z)   # каркас поверхности
plt.show()
````

`111` в `ax = fig.add_subplot(111, projection='3d')` — это закодированное сокращение для сетки подграфиков (subplots).

Это три цифры, которые означают:

- 1-я цифра — количество строк в сетке

- 2-я цифра — количество столбцов в сетке

- 3-я цифра — индекс текущего графика (нумерация по строкам слева направо)

То есть 111 = 1 строка, 1 столбец, 1-й (единственный) график.

```python
# Примеры с несколькими графиками:
# Сетка 2x2 (4 графика)
ax1 = fig.add_subplot(221)  # 1-й график (строка 1, столбец 1)
ax2 = fig.add_subplot(222)  # 2-й график (строка 1, столбец 2)
ax3 = fig.add_subplot(223)  # 3-й график (строка 2, столбец 1)
ax4 = fig.add_subplot(224)  # 4-й график (строка 2, столбец 2)

# Сетка 2x3 (6 графиков)

```

**Замечание.** Третья цифра (индекс) не может быть двузначной! То есть если у вас 10 и более графиков, нужно использовать явную форму:

##### Параметрическая поверхность $x=f(u,v),y=g(u,v),z=h(u,v)$:

````python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 50)
v = np.linspace(0, np.pi, 50)
U, V = np.meshgrid(u, v)

X = np.sin(V) * np.cos(U)
Y = np.sin(V) * np.sin(U)
Z = np.cos(V)

ax.plot_surface(X, Y, Z)
plt.show()
````

##### Трехмерная линия (параметрическая кривая)

````python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

t = np.linspace(0, 4*np.pi, 200)
x = np.sin(t)
y = np.cos(t)
z = t

ax.plot(x, y, z)

plt.show()
````

## Построение таблиц

### Базовые способы

Рассмотрим базовые способы вывода таблиц. Обычный вывод таблицы можно осуществить с помощью печати текста с табуляцией в консоль.

````python
# print() и f-строки
data = [["Имя", "Возраст", "Город"], 
        ["Анна", 28, "Москва"], 
        ["Павел", 34, "СПб"]]
for row in data:
    print(f"{row[0]:<10} {row[1]:<8} {row[2]:<10}")
````

Можно модифицировать программу для того, чтобы в цикле формировался код MD, который можно интерпретировать непостредственно в JupyterLab.

````python
# print() и f-строки (MD таблица)
data = [["Имя", "Возраст", "Город"], 
        ["Анна", 28, "Москва"], 
        ["Павел", 34, "СПб"]]

# Выводим заголовок с разделителем
header = data[0]
print(f"| {header[0]} | {header[1]} | {header[2]} |")
print(f"|{'---'}|{'---'}|{'---'}|")

# Выводим данные
for row in data[1:]:
    print(f"| {row[0]} | {row[1]} | {row[2]} |")
````
