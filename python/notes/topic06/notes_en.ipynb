#### Two-dimensional plot (`plt.plot`)

The `plt.plot` command plots two-dimensional plots. Examples of using the `plt.plot` function:

- 1. Explicitly defined functions $y=f(x)$.
```python
plt.plot(x, y, options)
```

- 2. Parametrically defined curves $x=f(t)$, $y=g(t)$.

```python
t = np.linspace(0, 2*np.pi, 100)
x = np.cos(t)
y = np.sin(t)
plt.plot(x, y, options)
```

- 3. Point sets.

```python
# Method 1 - List of coordinate pairs
plt.plot([[x1, x2, x3], [y1, y2, y3]], 'o', options)

# Method 2 - Separate X and Y arrays
plt.plot([x1, x2, x3], [y1, y2, y3], 'o', options)
```

- 4. Multiple curves on one plot.
```python
plt.plot(x1, y1, options1, x2, y2, options2, ...)
```

##### Main options (specified as keyword arguments)

| Option | Value (underlined by default) | Explanation |
|-------|--------------------------------------|-----------|
| `color` / `c` | `'b'` (blue), `'r'`, `'g'`, `'k'` (black), `'m'`, `'c'`, `'y'`, `'w'` | Line/marker color |
| `linestyle` / `ls` | `'-'` (solid), `'--'` (dash), `'-.'` (dash-dot), `':'` (dash), `''` (no line) | Line style |
| `marker` | `''` (none), `'o'`, `'s'`, `'^'`, `'D'`, `'*'`, `'+'`, `'x'`, `'.'` | Point marker type |
| `linewidth` / `lw` | `1.0` | Line width in points |
| `markersize` / `ms` | `6.0` | Marker size |
| `alpha` | `1.0` (opaque) | Transparency (0–1) |
| `label` | `''` | Legend label |

````python
# Data generation
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x + 0.5) * 0.8
y4 = np.cos(x + 0.5) * 0.8

# Creating a figure and axes
plt.figure(figsize=(10, 6))

# Example 1: Blue solid line with round markers
plt.plot(x, y1, 
         color='b',           # blue color
         linestyle='-',       # solid line
         marker='o',          # round markers
         linewidth=2.0,       # line thickness 2 points
         markersize=8.0,      # marker size 8
         alpha=0.8,           # transparency 80%
         label='sin(x)')

# Example 2: Red dashed line with square markers
plt.plot(x, y2, 
         color='r',           # red color
         linestyle='--',      # dashed line
         marker='s',          # square markers
         linewidth=1.5,       
         markersize=6.0,
         alpha=0.9,
         label='cos(x)')

# Example 3: Green dashed line with triangle markers
plt.plot(x, y3, 
         color='g',           
         linestyle='-.',      # dash-dot line
         marker='^',          # triangle markers
         linewidth=2.5,
         markersize=9.0,
         alpha=0.7,
         label='sin(x+0.5)*0.8')

# Example 4: Purple dotted line with diamond markers
plt.plot(x, y4, 
         color='m',           
         linestyle=':',       # dotted line
         marker='D',          # diamond-shaped markers
         linewidth=2.0,
         markersize=7.0,
         alpha=0.6,
         label='cos(x+0.5)*0.8')

# Adding axis title and labels
plt.title('An example of using the basic matplotlib options', fontsize=14)
plt.xlabel('X', fontsize=12)
plt.ylabel('Y', fontsize=12)

# Adding a grid
plt.grid(True, alpha=0.3)

# Adding a legend
plt.legend(loc='upper right', fontsize=10)

# Displaying a graph
plt.tight_layout()
plt.show()
````

#### Polar Coordinate System

For a polar plot, consider the following example:

````python
theta = np.linspace(0, 2*np.pi, 100)
r = 1 + np.cos(theta)

plt.axes(projection='polar')
plt.plot(theta, r, color="red")
````

#### Three-dimensional graph (`plot3d`)

For 3D graphics, the `mpl_toolkits.mplot3d` module is used.

##### Explicit function $z=f(x,y)$

````python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax.plot_surface(X, Y, Z)   # surface
plt.show()
````

````python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax.plot_wireframe(X, Y, Z)   # surface frame
plt.show()
````

`111` in `ax = fig.add_subplot(111, projection='3d')` is a coded shorthand for a grid of subplots.

These three digits represent:

- 1st digit — number of rows in the grid

- 2nd digit — number of columns in the grid

- 3rd digit — index of the current chart (numbered by rows from left to right)

That is, 111 = 1 row, 1 column, 1st (only) chart.

```python
# Examples with multiple plots:
# 2x2 grid (4 plots)
ax1 = fig.add_subplot(221)   # 1st plot (row 1, column 1)
ax2 = fig.add_subplot(222)   # 2nd plot (row 1, column 2)
ax3 = fig.add_subplot(223)   # 3rd plot (row 2, column 1)
ax4 = fig.add_subplot(224)   # 4th plot (row 2, column 2)

# 2x3 grid (6 plots)
ax1 = fig.add_subplot(231)   # row 1, column 1
ax2 = fig.add_subplot(232)   # row 1, column 2
ax3 = fig.add_subplot(233)   # row 1, column 3
ax4 = fig.add_subplot(234)   # row 2, column 1
ax5 = fig.add_subplot(235)   # row 2, column 2
ax6 = fig.add_subplot(236)   # row 2, column 3
```

##### Parametric surface $x=f(u,v),y=g(u,v),z=h(u,v)$:

````python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 50)
v = np.linspace(0, np.pi, 50)
U, V = np.meshgrid(u, v)

X = np.sin(V) * np.cos(U)
Y = np.sin(V) * np.sin(U)
Z = np.cos(V)

ax.plot_surface(X, Y, Z)
plt.show()
````

##### Three-dimensional line (parametric curve)

````python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

t = np.linspace(0, 4*np.pi, 200)
x = np.sin(t)
y = np.cos(t)
z = t

ax.plot(x, y, z)

plt.show()
````

## Table Construction

### Basic Methods

Let's look at basic methods for displaying tables. Standard table display can be achieved by printing tab-delimited text to the console.

````python
# print() and f-strings
data = [["Name", "Age", "City"],
        ["Anna", 28, "Moscow"],
        ["Pavel", 34, "St. Petersburg"]]
for row in data:
    print(f"{row[0]:<10} {row[1]:<8} {row[2]:<10}")
````

The program can be modified to generate MD code in the loop that can be interpreted directly in JupyterLab.

````python
# print() and f-rows (MD table)
data = [["Name", "Age", "City"],
        ["Anna", 28, "Moscow"],
        ["Pavel", 34, "St. Petersburg"]]

# Print the header with a separator
header = data[0]
print(f"| {header[0]} | {header[1]} | {header[2]} |")
print(f"|{'---'}|{'---'}|{'---'}|")

# Print the data
for row in data[1:]:
    print(f"| {row[0]} | {row[1]} | {row[2]} |")
````

### Working with the pandas library and the concept of a DataFrame

#### Displaying a DataFrame

````python
import pandas as pd
df = pd.DataFrame(data)
print(df)     # basic output
df.head(10)   # first rows
df.tail()     # last
````

The first line becomes the headings:

````python
df = pd.DataFrame(data[1:], columns=data[0])
df.tail()
````

#### Table Styling

````python
# Style only numeric columns
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Apply styles only to numeric columns
df.style.background_gradient(cmap='Blues', subset=numeric_cols)
df.style.bar(subset=['Возраст'], color='#5fba7d')
df.style.highlight_max(color='lightgreen', subset=numeric_cols)
````
