### Type conversion (explicit and implicit)

#### Implicit (automatic)

Python automatically changes the type to avoid data loss. It always goes up (from simple to complex):
int → float → complex

````python
x = 10         # int
y = 3.5        # float
z = x + y      # the result will be float(13.5)
````

````python
int("123")      # 123
float("4.5")    # 4.5
str(100)        # "100"
int(7.9)        # 7 (just discards the .9)
bool(0)         # False (empty and zero → False)
````

Important: Explicit conversion may cause an error (ValueError) if the string is not a number:

### IEEE 754
The `double` data type in Python, as in other programming languages, corresponds to a double-precision floating-point number, which is governed by the [IEEE 754](https://en.wikipedia.org/wiki/IEEE_floating_point) standard, which was published in **1985**. Before that, computers handled floating-point numbers in different ways!

IEEE 754 contains the following clauses:
- Floating-point number representation: $(-1)^s \times c \times b^q$.
- Two infinities $+\infty$ and $-\infty$.
- Two types of **NaN**: "quiet" NaN (**qNaN**) and signaling NaN (**sNaN**).
- **qNaN** does not throw an exception at the floating point unit (FPU) level until you check the result of the calculation.
- **sNaN** throws an exception from the FPU if you use this value in calculations. This NaN type can be useful for initialization.
- **Rounding** rules.
- Rules for operations like $\frac{0}{0}, \frac{1}{-0}, \ldots$

#### Example 4. Two infinities

If we want to avoid the error and work with infinities,
we can use the `math` module or Python's built-in infinity support:

````python
x = -1.0
y = 0.0

if y == 0.0:
    result = float('inf') if x > 0 else float('-inf')
else:
    result = x / y

print(result)
````

#### Example 5. Nan's

To illustrate a quiet NaN, we'll use the standard NaN value, which is created using float('nan'). This NaN doesn't throw exceptions during operations and is passed into the calculation result.

````python
qnan = float('nan')

# Operation with qNaN does not throw an exception, it just propagates it
result = qnan + 1
print(result)  # Print 'nan'
````

````python
# Create sNaN (signaling NaN)
snan = float('nan')

# Function to check for NaN and throw an exception
def check_and_compute(value):
    if str(value) == 'nan':  # Check if the value is NaN
        raise ValueError("Signaling NaN (sNaN) is used, causing an exception")
    return value + 1

# Example of use with sNaN
try:
    result = check_and_compute(snan)
except ValueError as e:
    print(e)  # Prints: Signaling NaN (sNaN) used, throwing exception
````

### Sets
Sets are unordered collections of unique elements. They enable set theory operations such as union, intersection, and difference.

````python
# Empty set
empty_set = set()

# Set with elements
numbers = {1, 2, 3, 4, 5}

# Creating a set from a list
list_set = set([1, 2, 3, 4, 4, 5]) # Duplicates are removed

print(list_set)
````

#### Basic operations

- Adding and removing elements

````python
s = {1, 2, 3}
s.add(4)     # Adds an element
print(s)
s.remove(2)  # Removes an element (if it doesn't exist, raises an error)
print(s)
s.discard(5) # Removes an element if it exists, but doesn't raise an error
print(s)
s.clear()    # Clears the set
print(s)
````

- Operations on sets

````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Union (A ∪ B)
union_set = A | B
print(union_set)

# Intersection (A ∩ B)
intersection_set = A & B
print(intersection_set)

# Difference (A \ B)
difference_set = A - B
print(difference_set)

# Symmetric Difference (A △ B)
symmetric_difference_set = A ^ B
print(symmetric_difference_set)
````

- Verification of affiliation

````python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

print(1 in A)           # True
print(A.issubset(B))    # True (A ⊆ B)
print(B.issuperset(A))  # True (B ⊇ A)
````

### Nested Structures

Nested structures are data structures (lists, dictionaries, tuples, sets) that contain other data structures. This allows for the creation of complex hierarchical models (matrices, catalogs, in-memory databases).

#### Matrices

A list whose elements are other lists.

````python
# Creating a 3x3 matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Access to element: [external_index][internal_index]
print(matrix[0][1])  # 2 (first row, second column)
print(matrix[2][2])  # 9

# Traversal (nested loops)
for row in matrix:
    for element in row:
        print(element, end=' ')
    print()  # New line break
````

#### Nested Dictionaries

A dictionary whose values ​​are other dictionaries. Ideal for storing records with multiple fields.

````python
# Database of students
students = {
    "ivanov": {
        "name": "Ivan",
        "age": 20,
        "grades": [5, 4, 5, 3]
    },
    "petrova": {
        "name": "Maria",
        "age": 19,
        "grades": [5, 5, 5, 4]
    }
}

# Data access
print(students["ivanov"]["name"])       # Ivan
print(students["petrova"]["grades"][0]) # 5 (Maria's first assessment)

# Adding a new student
students["sidorov"] = {"name": "Peter", "age": 21, "grades": [4, 4, 3]}

# Safe access (to avoid KeyError)
age = students.get("petrova", {}).get("age")  # 19
````

#### Dictionary List

The most common format for data exchange (JSON, API, parsing). Each object is a separate dictionary within a shared list.

````python
# List of items in the cart
cart = [
    {"id": 1, "name": "Laptop", "price": 1500},
    {"id": 2, "name": "Mouse", "price": 25},
    {"id": 3, "name": "Keyboard", "price": 80}
]

# Search for a product by condition (loop + condition)
for item in cart:
    if item["price"] > 100:
        print(f"Expensive goods: {item['name']}")  # Ноутбук

#Price summation (cycle + accumulator)
total = sum(item["price"] for item in cart)  # 1605
````

#### Complex combination (List of dictionaries with nested lists)

A real case from the life of a developer.

````python
# Online store order data
orders = [
    {
        "order_id": 101,
        "client": "Anna",
        "items": [
            {"name": "Book", "qty": 2, "price": 500},
            {"name": "Pen", "qty": 5, "price": 50}
        ]
    },
    {
        "order_id": 102,
        "client": "Boris",
        "items": [
            {"name": "Laptop", "qty": 1, "price": 120000}
        ]
    }
]

# How to extract the price of the second item in the first order?
price = orders[0]["items"][1]["price"]  # 50

# Calculate the total cost of each order (nested loops)
for order in orders:
    total = 0
    for product in order["items"]:
        total += product["qty"] * product["price"]
    print(f"Order {order['order_id']} for the amount of: {total} bucks.")
````

#### 6. Subtleties and nuances (Important!)

| Situation | Solution |
| :--- | :--- |
| **Copying Nested Structures** | `copy.copy()` (shallow) copies only the outer layer, while inner lists/dictionaries remain shared. Use `copy.deepcopy()` for a full, independent copy. |
| **Checking for Key Existence** | Don't write `if dict["key"]` - you'll get a `KeyError`. Use `if "key" in dict` or the `.get()` method. |
| **Deep Access** | Chains like `a["b"][0]["c"]` are dangerous. If an error occurs at any level, the program will crash. Use `try/except` or the `jmespath` library for complex queries. |
| **Iterating Over Deep Structures** | To traverse unknown depths, use **recursive functions** (a function that calls itself). |
