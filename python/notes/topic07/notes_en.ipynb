## Concept of Infinitesimal and Infinitely Large Quantities

### Infinitesimal and Infinitely Large Quantities

**Definition 1.** A function $f(x)$ is called an infinitesimal if
$$\lim_{x \to 0} f(x) = 0 \tag{1}$$

Examples of infinitesimal quantities: $x, \sqrt{x}, x^2, x^3, x^4, \sin(x), \cos(x), -1, \operatorname{tg}(x)$.

In standard coordinates, the graphs of infinitesimal quantities near zero are practically indistinguishable from each other, making it impossible to perform a visual analysis of the functions.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-2, 2, 100)

plt.plot(x, x, label='x')
plt.plot(x, np.sqrt(np.maximum(x, 0)), label='sqrt(x)')
plt.plot(x, x**2, label='x^2')
plt.plot(x, x**3, label='x^3')
plt.plot(x, x**4, label='x^4')
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.plot(x, -np.ones_like(x), label='-1')
plt.plot(x, np.tan(x), label='tg(x)')

plt.legend()
plt.grid()
plt.show()
```

**Definition 2.** The order of smallness of an infinitesimal quantity $f(x)$ is the smallest positive number $p$ for which the limit
$$b = \lim_{x \to 0} \frac{f(x)}{x^p}$$
exists and is nonzero.
Examples of determining orders of smallness:
1) The infinitesimal quantity $y = x^2$ has the second order of smallness, since
$$\lim_{x \to 0} \frac{f(x)}{x^2} = \lim_{x \to 0} \frac{x^2}{x^2} = 1 \neq 0.$$

2) The infinitesimal quantity $y = \sqrt{x}$ has the order of smallness $\frac{1}{2}$, since
$$\lim_{x \to 0} \frac{\sqrt{x}}{x^{1/2}} = 1 \neq 0.$$

3) The infinitesimal quantity $y = x$ has the order of smallness $1$, since
$$\lim_{x \to 0} \frac{x}{x} = 1 \neq 0.$$

**Definition 3.** A function $F(x)$ is called an infinitely large quantity if
$$\lim_{x \to \infty} F(x) = \infty.$$

The following relationship exists between an infinitely small quantity $f(x)$ and an infinitely large quantity $F(x)$:
$$f(x) = \frac{1}{F(1/x)} \text{ is an infinitely small quantity,}$$
$$F(x) = \frac{1}{f(1/x)} \text{ is an infinitely large quantity.}$$

Infinitesimal quantities can appear in various cases:
1) Definition using some analytical formula.<br>
2) For any continuous function $g(x)$ and any point $x_0$, the quantity
$$f(x) = g(x + x_0) - g(x_0)$$
is infinitely small small quantity.<br>
3) Error of **numerical methods**.

### Logarithmic scale.

Using a natural scale to plot graphs of infinitesimal quantities does not allow us to consider the behavior in a small neighborhood of zero. The explanation for this is obvious: formula (1) serves this fact. We will use a logarithmic scale for both coordinate axes to plot the graph of an infinitesimal quantity. To do this, we take the logarithm of the expression $y = f(x)$ to base 10. We obtain

$$\lg y = \lg f(x)\tag{2}.$$
Let's assume that the right-hand side of expression (2) can be expressed in terms of $\lg x$:
$$\lg f(x) = g (\lg x),\tag{3}$$
where $g(x)$ is some analytic function. Let $\bar{x} = \lg x$, $\bar{y} = \lg y$, then
equation (3) takes the form
$$\bar{y} = g(\bar{x}).\tag{4}$$
The graph of function (4) will be called the $\lg-\lg$ graph of the original function $f(x)$

```python
plt.plot(x, x, label='x')
plt.plot(x, np.sqrt(np.maximum(x, 0)), label='sqrt(x)')
plt.plot(x, x**2, label='x^2')
plt.plot(x, x**3, label='x^3')
plt.plot(x, x**4, label='x^4')
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.plot(x, -np.ones_like(x), label='-1')
plt.plot(x, np.tan(x), label='tg(x)')

# plt.xscale('log')  # logarithmic scale for x-axis
plt.yscale('log')    # logarithmic scale for y-axis

plt.legend()
plt.grid()
plt.show()
```

### Operations for lg-lg graphs

We will call the function $g(x) = \lg f(x)$ the $\lg$ image of the function $f(x)$, and represent this correspondence as
$$g(\tilde{x}) \leftrightarrow f(x).$$
Suppose we have two infinitesimal quantities $f_1(x)$ and $f_2(x)$. Let's construct their $\lg$-images:
$$\tilde{y} = g_2(\tilde{x}) = \lg f_1(x),$$
$$\tilde{y} = g_2(\tilde{x}) = \lg f_2(x).$$
Arithmetic operations can be performed on $\lg$-images:
$$c g_1(\tilde{x}) = c \lg f_1(x) = \lg f_1^c(x),$$
which implies
$$c g_1(\tilde{x}) \leftrightarrow f_1^c(x).$$
Similarly, we obtain
$$c + g_1(\tilde{x}) \leftrightarrow 10^c \cdot f_1(x),$$
$$g_1(\tilde{x}) + g_2(\tilde{x}) \leftrightarrow f_1(x) \cdot f_2(x),$$
$$g_1(\tilde{x}) \cdot g_2(\tilde{x}) \leftrightarrow f_2(x)^{f_1(x)}.$$

Conclusion: Arithmetic operations on logarithmic images of infinitesimal functions correspond to certain operations on the original infinitesimals.
A similar situation occurs with the integral Fourier transform.

When analyzing numerical methods, determining the order of accuracy of these methods plays an important role. For example, the order of accuracy for calculating a definite integral using the trapezoidal method is 2. This means that decreasing the integration step by a factor of 10 reduces the method's error by a factor of 100. Determining the order of accuracy of a method is a rather complex mathematical problem. We will demonstrate how this can be accomplished using our approach.

Let's find the accuracy of two methods for calculating the derivative $f'(x_0)$.

1) Differentiation using the right-hand formula
$$f'(x_0) \approx \frac{f(x_0 + h) - f(x_0)}{h}, \tag{10}$$
where the differentiation step h is theoretically an infinitesimal value.
The total error function of formula (10) has the form
$$\varphi(h) = |f'(x_0) - \frac{f(x_0 + h) - f(x_0)}{h}| \tag{11}$$
It is known that $ \varphi(h) \to 0 $, i.e., $ \varphi $ is an infinitesimal quantity.
Let us construct the $ \lg h $-graph of the error (11). To find the asymptote of the $ \lg h $-graph, we find two points. From the theory of numerical methods, it is known that formula (10) is of the first order of smallness, i.e., $ \varphi(h) \approx C \cdot h $. Hence, $ \lg(\varphi(h)) \approx \lg(C \cdot h) = \lg(C) + \lg(h) $, i.e., the slope of the oblique asymptote must be equal to $ k = 1 $.
Let's test this reasoning with an example.

## Time Complexity of Algorithms

### Basic Concepts

**Time complexity** is the number of elementary operations performed by an algorithm depending on the input data size $ n $.

Notation: $ T(n) $ is the running time of the algorithm for input data of size $ n $.

### Asymptotic Notations

Asymptotic notations are used to describe the behavior of a function $ T(n) $ as $ n \to \infty $.

#### Big O — $ O(f(n)) $

**Definition:** $ T(n) = O(f(n)) $ if there exist constants $ C > 0 $ and $ n_0 > 0 $ such that for all $ n \ge n_0 $ the following holds:
$$
T(n) \le C \cdot f(n).
$$

**Meaning:** $ f(n) $ is the **upper bound** on the running time (the algorithm runs no slower than $ f(n) $).

Used to estimate the **worst case**.

#### "Big Omega" — $ \Omega(f(n)) $

**Definition:** $ T(n) = \Omega(f(n)) $ if there exist constants $ C > 0 $ and $ n_0 > 0 $ such that for all $ n \ge n_0 $ the following holds:
$$
T(n) \ge C \cdot f(n).
$$

**Meaning:** $ f(n) $ is the **lower bound** on the running time (the algorithm runs no faster than $ f(n) $).

Used to estimate the **best case**.

#### "Big Theta" — $ \Theta(f(n)) $

**Definition:** $ T(n) = \Theta(f(n)) $ if both of the following are true:
$$
T(n) = O(f(n)) \quad \text{and} \quad T(n) = \Omega(f(n)).
$$

**Meaning:** $ f(n) $ is the **exact asymptotic bound** (the algorithm grows as $ f(n) $ up to a constant).

### Complexity Class Examples

| Class | Notation | Algorithm Example | Characteristic |
| ---|--- |--- |--- |
| **Constant** | $ O(1) $ | Access to an array element by index | Time does not depend on $ n $ |
| **Logarithmic** | $ O(\log n) $ | Binary Search | Each step halves the problem |
| **Linear** | $ O(n) $ | Maximum Search, Linear Search | Proportional to $ n $ |
| **Linear-Logarithmic** | $ O(n \log n) $ | Merge Sort, QuickSort (average case) | Efficient Sort |
| **Quadratic** | $ O(n^2) $ | Bubble Sort, Insertion Sort (worst case) | Double Loop |
| **Cubic** | $ O(n^3) $ | Naive Matrix Multiplication | Triple Loop |
| **Exponential** | $ O(2^n) $ | Enumerate all subsets | Explosive growth |
| **Factorial** | $ O(n!) $ | Enumerate all permutations | Slowest class |

### Comparison of Algorithms by Runtime

#### Example 1. Merge Sort and Quick Sort

| Algorithm | Best Case | Average Case | Worst Case | Memory |
| --- | --- |--- |--- |--- |
| **Merge Sort** | $ O(n \log n) $ | $ O(n \log n) $ | $ O(n \log n) $ | $ O(n) $ |
| **Quick Sort** | $ O(n \log n) $ | $ O(n \log n) $ | $ O(n^2) $ | $ O(\log n) $ |

**Conclusions:**
- Merge Sort **guarantees** $ O(n \log n) $ in all cases, but requires additional $ O(n) $ memory.
- Quicksort is **on average** faster due to a better constant and less memory usage, but in the worst case (poor choice of pivot) it degrades to $ O(n^2) $.

```python
# 1. Merge Sort - O(n log n)
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# 2. Quick Sort - O(n log n) on average
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    return quick_sort(left) + middle + quick_sort(right)
```

```python
test_arr = [64, 34, 25, 12, 22, 11, 90]

print("Merge Sort:     ", merge_sort(test_arr.copy()))
print("Quick Sort:     ", quick_sort(test_arr.copy()))
```

#### Example 2. Linear Search and Binary Search

| Algorithm | Time | Condition |
|--- |--- | ---|
| **Linear Search** | $ O(n) $ | Array is unsorted |
| **Binary Search** | $ O(\log n) $ | Array is sorted |

**Conclusion:** For large $ n $, binary search is significantly faster, but requires pre-sorting (which costs $ O(n \log n) $).

```python
# Linear search - works on any array
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Binary search - works only on sorted arrays
def binary_search(arr, target):
    left = 0
    right = len(arr) - 1
    
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

```python
test_arr = [64, 34, 25, 12, 22, 11, 90]
sorted_arr = sorted(test_arr) # [11, 12, 22, 25, 34, 64, 90]

target = 25

print(f"Array: {test_arr}")
print(f"Sorted: {sorted_arr}")
print(f"Searching: {target}\n")

# Linear Search
pos = linear_search(test_arr, target)
print(f"Linear Search: position {pos}")

# Binary Search
pos = binary_search(sorted_arr, target)
print(f"Binary Search: position {pos}")
```
