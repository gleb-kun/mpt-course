# Тема #2. Контейнеризация

## Виртуализация vs Контейнеризация:

- Виртуальная машина (VM): Эмулирует железо, имеет свою ОС (гостевую). Тяжелая, долгий старт.

- Контейнер (Docker): Использует ядро хостовой ОС (Linux). Легкий, запускается за миллисекунды. Это не виртуализация, а изоляция процессов (используются пространства имен (namespaces) и контрольные группы (cgroups) ядра Linux).

## Простейший пример использования Docker


1. **Шаг 1.** Получаем образ Ubuntu

```bash
docker pull ubuntu:22.04
```

2. **Шаг 2.** Создаём контейнер

Создадим контейнер с интерактивным режимом и именем my-ubuntu:

```bash
docker create -it --name my-ubuntu ubuntu:22.04
```

3. **Шаг 3.** Запускаем и попадаем внутрь!

```bash
docker start -ai my-ubuntu
```

- `-a` (attach) — прикрепляем наш текущий терминал к терминалу контейнера.

- `-i` (interactive) — разрешаем ввод команд.

4. **Шаг 4.** Если мы просто запустили контейнер, но не зашли, отдельно можно зайти в него так:

```bash
docker exec -it my-ubuntu bash
```

5. **Шаг 5.** Далее работаем внутри контейнера как Вам угодно.
