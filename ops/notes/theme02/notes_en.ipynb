# Topic #2. Containerization

## Virtualization vs. Containerization:

- Virtual Machine (VM): Emulates hardware and has its own OS (guest). Heavy, slow startup.

- Container (Docker): Uses the host OS kernel (Linux). Lightweight, starts in milliseconds. This isn't virtualization, but rather process isolation (using Linux kernel namespaces and control groups).

## A simple example of using Docker

1. **Step 1.** Get the Ubuntu image

```bash
docker pull ubuntu:22.04
```

2. **Step 2.** Create a container

Create a container in interactive mode named my-ubuntu:

```bash
docker create -it --name my-ubuntu ubuntu:22.04
```

3. **Step 3.** Run and get inside!

```bash
docker start -ai my-ubuntu
```

- `-a` (attach) — attach our current terminal to the container's terminal.

- `-i` (interactive) — allow command entry.

4. **Step 4.** If we simply launched the container but didn't log in, we can log in separately like this:

```bash
docker exec -it my-ubuntu bash
```

5. **Step 5.** Next, work within the container as you wish.
