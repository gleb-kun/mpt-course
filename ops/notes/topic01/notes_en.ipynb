# Topic #1. Version control systems

## Types of Version Control Systems

| System Type | Key Feature | Examples |
|-------------|----------------------|---------|
| Local (Local VCS) | Stores the change history on the local computer. This is the simplest and most outdated approach. | RCS (Revision Control System), SCCS |
| Centralized (CVCS) | Uses a single central server where the entire project history is stored. Developers retrieve files from the server and push changes back. | Subversion (SVN), Perforce, CVS (outdated) |
| Distributed (DVCS) | Each developer has a complete local copy of the entire repository (including the entire history). This allows for offline work and makes the system more reliable. | Git, Mercurial, Bazaar |

## Git

Git is a distributed system that has become the undisputed standard in modern development. According to surveys, over 93% of developers use it. Git is the foundation of popular platforms such as GitHub, GitLab, and Bitbucket, which provide repository hosting and powerful tools for collaboration and CI/CD.

### Essential Daily Commands

A minimal set of commands that, statistically, covers 80% of all daily tasks:

| Command | What it does | Example |
| :--- | :--- | :--- |
| **`git init`** | Creates a new Git repository in the current folder | `git init` |
| **`git clone <url>`** | Downloads a remote repository to your computer | `git clone https://github.com/user/repo.git` |
| **`git status`** | Checks the status: which files have changed, are added to the index, or are not tracked | `git status` |
| **`git add <file>`** | Adds changes to the index (staging area) | `git add README.md` or `git add .` (add all) |
| **`git commit -m "message"`** | Creates a commit (snapshot) with the changes added to the stage | `git commit -m "Added README"` |
| **`git log`** | Shows the commit history | `git log --oneline` (short view) |
| **`git push`** | Pushes your local commits to a remote repository (e.g., GitHub) | `git push origin main` |
| **`git pull`** | Fetches changes from the remote repository and merges them into your current branch | `git pull` |

### .git folder tree

The structure of the `.git` directory (located in the root of the repository):

```bash
.git/
├── HEAD                    # (!) Points to the current branch (for example: ref: refs/heads/main)
├── config                  # (!) Configuration of this repository (remote repository addresses, branch settings)
├── description             # (!) Description of the repository (used for GitWeb)
├── index                   # A snapshot of files in the index (staging area) that will be added to the next commit
├── branches/               # [DEPRECATED] Previously used to store references to remote branches
├── hooks/                  # Handler scripts (hooks) that are triggered for specific events (e.g., before a commit)
│   └── (pre-commit.sample, commit-msg.sample, etc.)
├── info/                   # Additional repository information
│   └── exclude             # List of ignored files for this repository (similar to .gitignore, but not included in commits)
├── logs/                   # A log of all ref changes (Reflog) — a "rollback log"
│   └── refs/heads/         # Local branch change history
│   └── (main, feature, etc.)
├── objects/                # (!) Git DATABASE (all objects: commits, trees, files)
│   ├── pack/               # Packed (compressed) objects to save space and speed up work
│   ├── info/               # Service information about packed objects
│   └── [0-9a-f][0-9a-f]/   # 256 folders named by the first two characters of the object's SHA-1 hash
│   └── (objects in folders, compressed and named by the rest of the hash)
├── refs/                   # References to objects (branches, tags, remote branches)
│   ├── heads/              # References to local branches (each branch is a file with the hash of the latest commit)
│   │   └── main, dev, feature...
│   ├── remotes/            # References to remote branches (snapshots of the state of remote repositories)
│   │   └── origin/         # (e.g., origin/main, origin/dev)
│   └── tags/               # References to tags (usually to specific commits, immutable labels)
│   └── v1.0.0, release...
├── ORIG_HEAD               # Stores a reference to the previous state of HEAD (used when undoing merge and rebase operations)
├── FETCH_HEAD              # Stores information about branches retrieved during the last git fetch command
└── MERGE_HEAD / MERGE_MODE # Temporary files that appear during conflict resolution mergers
```

### `.gitignore` file

The `.gitignore` file is a special text file that tells Git which files and folders should not be tracked and added to the repository.

#### Syntax and Writing Rules

| Rule | Example | What does it mean |
| :--- | :--- | :--- |
| **Empty line** | (whitespace) | Ignored |
| **Comment** | `# This is a comment` | Ignored |
| **Filename** | `secret.txt` | Ignores the file `secret.txt` in any folder |
| **Folder** | `logs/` | Ignores the entire `logs` folder (and everything inside) |
| **The `*`** character | `*.log` | Ignores all files with the `.log` extension |
| **Character `?`** | `file?.txt` | Ignores `file1.txt`, `fileA.txt`, etc. (any single character) |
| **Range `[abc]`** | `file[0-9].txt` | Ignores `file0.txt` ... `file9.txt` |
| **`**` (two asterisks)** | `**/temp/*.tmp` | Ignores all `.tmp` in the `temp` folder at any depth |
| **Exclamation mark `!`** | `!important.log` | **Undoes** ignoring `important.log` (if it was `*.log`) |
| **Trailing slash `/`** | `build/` | Ignores **only the `build` folder** |
| **Leading slash `/`** | `/target/` | Ignores `target` **only in the project root** |

#### Example of a `.gitignore` gitignore file

```bash
# Comment: Ignore system files
.DS_Store
Thumbs.db

# Ignore the dependencies folder (Node.js)
node_modules/

# Ignore all log files
*.log
logs/

# Ignore temporary Python files
__pycache__/
*.pyc

# Ignore files with the environment and secrets
.env
.env.local
*.key
*.pem

# Ignore IDE files
.idea/
.vscode/
*.iml

# BUT! We want to monitor one important log file (we'll override the *.log rule)
!important.log
```

### Commit Graph

Git stores history not as a linear list, but as a Directed Acyclic Graph (DAG). Each commit is a node in this graph, and arrows from the parent commit(s).

#### Simple linear history (no branches)

```text
A → B → C → D
```

- `A` is the first commit (no parents)

- `B` is the parent of `A`

- `C` is the parent of `B`

- `D` is the parent of `C` (the current commit, HEAD)

Each commit (except the first) points to exactly one parent. This is the simplest case.

#### Branching

When you create a new branch, the graph splits:

```text
          E → F → G (feature branch)
         ↗
A → B → C → D (main branch)
         ↑
         └── branch point (commit C)
```

The main branch continues straight ahead: `A → B → C → D`.
The feature branch forked from commit `C` and went its own way: `C → E → F → G`.
Commit `E` has `C` as its parent (it knows where it came from).

#### Merge

When you merge the `feature` branch into `main`, a merge commit is created with two parents:
```
A → B → C → D → M (main)
         ↘     ↗
          E → F → G (feature)
```

Commit `M` (the merge commit) has two parents: `D` (from main) and `F` (from feature).
This shows that `M` combines changes from both branches.
The graph stops being a simple tree and becomes a directed acyclic graph.

#### What the graph looks like in reality (commands)

| Command | What it shows |
|---------|----------------|
| `git log --oneline` | Linear list of commits (doesn't show the graph) |
| `git log --oneline --graph` | Shows an ASCII graph of branches and merges |
| `git log --oneline --graph --all` | Shows a graph of all branches in the repository |
| `git log --oneline --graph --decorate` | Adds branch and tag names to commits |

Example output of `git log --oneline --graph`:
```text
* a1b2c3d (HEAD -> main) Merge branch 'feature'
|\
| * f9e8d7c (feature) Added a new feature
| * e5d4c3b Fixed a bug in the logic
|/
* d3c2b1a Main functionality is ready
* c4b3a2z First commit
```

#### Terms and Concepts

| Term | Description |
|--------|---------|
| Parent | The previous commit from which the current commit is based |
| Root commit | A commit without parents (the very first one in the repository) |
| Branch | A movable pointer to one of the commits (usually the last one in the chain) |
| HEAD | A pointer to the current commit (the branch/commit you are currently on) |
| Branch point | The commit from which the new branch was forked |
| Merge commit | A commit with two (or more) parents; created when merging branches |
| Fast-forward merge | A special case where a branch can simply be "fast-forwarded" without creating a merge commit |

#### Fast-forward merge (special case)

If the feature branch hasn't diverged from main (i.e., main hasn't moved forward), the merge occurs without creating a separate commit:
```
Before:
main:   A ← B ← C
feature:        ↑
                └── D ← E

After git merge (fast-forward):
main:      A ← B ← C ← D ← E
feature:                   ↑
```

The main pointer is simply moved to the same commit as feature. The graph remains linear.
