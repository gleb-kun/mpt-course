# Тема #1. Системы контроля версий

## Типы систем контроля версий

| Тип системы | Ключевая особенность | Примеры |
|-------------|----------------------|---------|
| Локальные (Local VCS) | Хранят историю изменений на локальном компьютере. Это самый простой и устаревший подход. | RCS (Revision Control System), SCCS |
| Централизованные (CVCS) | Используют единый центральный сервер, где хранится вся история проекта. Разработчики получают файлы с сервера и отправляют изменения обратно. | Subversion (SVN), Perforce, CVS (устаревший) |
| Распределенные (DVCS) | Каждый разработчик имеет полную локальную копию всего репозитория (включая всю историю). Это позволяет работать офлайн и делает систему более надежной. | Git, Mercurial, Bazaar |

## Git

Современным лидером распределенных систем является программа Git.

Git — это распределенная система, которая стала безусловным стандартом в современной разработке. Согласно опросам, её используют более 93% разработчиков. Git лежит в основе популярных платформ: GitHub, GitLab и Bitbucket, которые предоставляют хостинг репозиториев и мощные инструменты для совместной работы и CI/CD.

### Установка

Windows: скачайте установщик с официального сайта `git-scm.com`.

macOS: используйте Homebrew: `brew install git`.

Linux (Ubuntu/Debian): `sudo apt update && sudo apt install git`, Arch-like: `sudo pacman -S git`.

### Основные команды на каждый день

Минимальный набор команд, который, по статистике, покрывает 80% всех ежедневных задач:

| Команда | Что делает | Пример |
| :--- | :--- | :--- |
| **`git init`** | Создаёт новый Git-репозиторий в текущей папке | `git init` |
| **`git clone <url>`** | Скачивает удалённый репозиторий к вам на компьютер | `git clone https://github.com/user/repo.git` |
| **`git status`** | Проверяет состояние: какие файлы изменены, добавлены в индекс или не отслеживаются | `git status` |
| **`git add <файл>`** | Добавляет изменения в индекс (staging area) | `git add README.md` или `git add .` (добавить всё) |
| **`git commit -m "сообщение"`** | Создаёт коммит (снимок) с добавленными в индекс изменениями | `git commit -m "Добавлен README"` |
| **`git log`** | Показывает историю коммитов | `git log --oneline` (краткий вид) |
| **`git push`** | Отправляет ваши локальные коммиты в удалённый репозиторий (например, на GitHub) | `git push origin main` |
| **`git pull`** | Забирает изменения из удалённого репозитория и сливает их с вашей текущей веткой | `git pull` |

### .git folder tree

Структура каталога `.git` (находится в корне репозитория):

```bash
.git/
├── HEAD                      # (!) Указывает на текущую ветку (например: ref: refs/heads/main)
├── config                    # (!) Конфигурация этого репозитория (адреса удалённых репозиториев, настройки веток)
├── description               # (!) Описание репозитория (используется для GitWeb)
├── index                     # Слепок (снимок) файлов в индексе (staging area), которые будут добавлены в следующий коммит
├── branches/                 # [УСТАРЕЛО] Раньше использовалось для хранения ссылок на удалённые ветки
├── hooks/                    # Сценарии-обработчики (хуки), которые срабатывают на определённые события (например, перед коммитом)
│   └── (pre-commit.sample, commit-msg.sample и др.)
├── info/                     # Дополнительная информация о репозитории
│   └── exclude               # Список игнорируемых файлов для этого репозитория (аналог .gitignore, но не попадает в коммиты)
├── logs/                     # Журнал всех изменений ссылок (Reflog) — это «журнал откатов»
│   └── refs/heads/           # История изменений локальных веток
│       └── (main, feature и т.д.)
├── objects/                  # (!) БАЗА ДАННЫХ Git (все объекты: коммиты, деревья, файлы)
│   ├── pack/                 # Упакованные (сжатые) объекты для экономии места и ускорения работы
│   ├── info/                 # Служебная информация об упакованных объектах
│   └── [0-9a-f][0-9a-f]/     # 256 папок, названных по первым двум символам хеша SHA-1 объекта
│       └── (объекты в папках, сжатые и именованные по оставшейся части хеша)
├── refs/                     # Ссылки на объекты (ветки, теги, удалённые ветки)
│   ├── heads/                # Ссылки на локальные ветки (каждая ветка — это файл с хешем последнего коммита)
│   │   └── main, dev, feature...
│   ├── remotes/              # Ссылки на удалённые ветки (слепки состояния удалённых репозиториев)
│   │   └── origin/           # (например, origin/main, origin/dev)
│   └── tags/                 # Ссылки на теги (чаще всего на конкретные коммиты, неизменяемые метки)
│       └── v1.0.0, release...
├── ORIG_HEAD                 # Хранит ссылку на предыдущее состояние HEAD (используется при отмене операций слияния и перебазирования)
├── FETCH_HEAD                # Хранит информацию о ветках, полученных во время последней команды git fetch
└── MERGE_HEAD / MERGE_MODE   # Временные файлы, которые появляются во время разрешения конфликтов слияния
```

### `.gitignore` файл

Файл `.gitignore` — это специальный текстовый файл, который указывает Git, какие файлы и папки не нужно отслеживать и добавлять в репозиторий.

#### Синтаксис и правила написания

| Правило | Пример | Что означает |
| :--- | :--- | :--- |
| **Пустая строка** | (пропуск) | Игнорируется |
| **Комментарий** | `# Это комментарий` | Игнорируется |
| **Имя файла** | `secret.txt` | Игнорирует файл `secret.txt` в любой папке |
| **Папка** | `logs/` | Игнорирует всю папку `logs` (и всё внутри) |
| **Символ `*`** | `*.log` | Игнорирует все файлы с расширением `.log` |
| **Символ `?`** | `file?.txt` | Игнорирует `file1.txt`, `fileA.txt` и т.д. (один любой символ) |
| **Диапазон `[abc]`** | `file[0-9].txt` | Игнорирует `file0.txt` ... `file9.txt` |
| **`**` (две звёздочки)** | `**/temp/*.tmp` | Игнорирует все `.tmp` в папке `temp` на любой глубине |
| **Восклицательный знак `!`** | `!important.log` | **Отменяет** игнорирование для `important.log` (если был `*.log`) |
| **Слеш `/` в конце** | `build/` | Игнорирует **только папку** `build` |
| **Слеш `/` в начале** | `/target/` | Игнорирует `target` **только в корне проекта** |

#### Пример тимпового файла `.gitignore`

```bash
# Комментарий: Игнорируем системные файлы
.DS_Store
Thumbs.db

# Игнорируем папку с зависимостями (Node.js)
node_modules/

# Игнорируем все файлы логов
*.log
logs/

# Игнорируем временные файлы Python
__pycache__/
*.pyc

# Игнорируем файлы с окружением и секретиками
.env
.env.local
*.key
*.pem

# Игнорируем файлы IDE
.idea/
.vscode/
*.iml

# НО! Хотим отслеживать один важный лог-файл (отменяем правило *.log)
!important.log
```

### Граф коммитов

Git хранит историю не как линейный список, а как направленный ациклический граф (Directed Acyclic Graph, DAG). Каждый коммит — это узел в этом графе, а стрелки нарисованы от родительского коммита (коммитов) к его дочернему коммиту (коммитам).

#### Простая линейная история (без веток)

```text
A → B → C → D
```

- `A` — первый коммит (нет родителей)

- `B` — родитель `A`

- `C` — родитель `B`

- `D` — родитель `C` (текущий коммит, HEAD)

Каждый коммит (кроме первого) указывает ровно на одного родителя. Это простейший случай.

#### Ветвление (Branching)

Когда вы создаёте новую ветку, граф раздваивается:

```text
          E → F → G   (ветка feature)
         ↗
A → B → C → D         (ветка main)
         ↑
         └── точка ветвления (коммит C)
```

Ветка main продолжает идти прямо: `A → B → C → D`.
Ветка feature ответвилась от коммита `C` и пошла своим путём: `C → E → F → G`.
У коммита `E` родитель — `C` (он знает, откуда он вырос).

#### Слияние (Merge)

Когда вы сливаете ветку `feature` в `main`, создаётся коммит слияния, у которого два родителя:
```
A → B → C → D → M   (main)
         ↘     ↗
          E → F → G (feature)
```

Коммит `M` (merge commit) имеет двух родителей: `D` (из main) и `F` (из feature).
Это показывает, что в `M` объединены изменения из обеих веток.
Граф перестаёт быть простым деревом и становится ориентированным ациклическим графом.

#### Как выглядит граф в реальности (команды)

| Команда | Что показывает |
|---------|----------------|
| `git log --oneline` | Линейный список коммитов (не показывает граф) |
| `git log --oneline --graph` | Показывает ASCII-граф ветвлений и слияний |
| `git log --oneline --graph --all` | Показывает граф всех веток в репозитории |
| `git log --oneline --graph --decorate` | Добавляет названия веток и тегов к коммитам |

Пример вывода `git log --oneline --graph`:
```text
*   a1b2c3d (HEAD -> main) Merge branch 'feature'
|\  
| * f9e8d7c (feature) Добавлена новая функция
| * e5d4c3b Исправлен баг в логике
|/  
* d3c2b1a Основной функционал готов
* c4b3a2z Первый коммит
```

#### Термины и понятия

| Термин | Описание |
|--------|----------|
| Родитель (parent) | Предыдущий коммит, на основе которого сделан текущий |
| Корневой коммит (root commit) | Коммит без родителей (самый первый в репозитории) |
| Ветка (branch) | Подвижный указатель на один из коммитов (обычно на последний в цепочке) |
| HEAD | Указатель на текущий коммит (ту ветку/коммит, где вы сейчас находитесь) |
| Точка ветвления | Коммит, от которого ответвилась новая ветка |
| Merge commit | Коммит с двумя (или более) родителями; создаётся при слиянии веток |
| Fast-forward merge | Особый случай, когда ветку можно просто "перемотать" вперёд без создания merge-коммита |

#### Fast-forward merge (особый случай)

Если ветка feature не расходилась с main (т.е. main не двигался вперёд), то слияние происходит без создания отдельного коммита:
```
До:
main:  A ← B ← C
feature:      ↑
              └── D ← E

После git merge (fast-forward):
main:  A ← B ← C ← D ← E
feature:              ↑
```

Просто перемещается указатель main на тот же коммит, что и feature. Граф остаётся линейным.
